#### Ex: How would we create a list of cube volumes for all cubes with side lengths between two integer values (inclusive)?
* create a function `cube_vol_list( low_side, high_side )` that returns the list of volumes
* print the result for each side length like: `side = 1 m, volume = 1 m^3`

````python
def cube_vol_list( lowside, highside ):

    # initialize list
    volumes = []


    # loop through all ints and build list
    for side in range(lowside, highside + 1):
        volumes.append(side ** 3)


    # return list
    return volumes

````

````python
# call function
myvols = cube_vol_list( 1, 5 )
print( myvols )
````

````python
# print nicely
side = 1

for vol in myvols:
    print( f"side = {side} m, volume = {vol} m^3" )
    side += 1
````

#### Ex: How would we build a list of inputted child ages?  
* Ask for ages until the user types a negative age.
* Display the number of children and the average age to 2 decimals

````python
# initialize list
ages = []

# get ages and build list
inputstr = input("Enter the ages of the children, separated by spaces: ")
for age in inputstr.split():
    ages.append( int(age) )

# calc stats
num_children = len(ages)
if num_children > 0:
    avg_age = sum(ages) / num_children
else:
    avg_age = 0


# output results
print(f"Child ages: {ages}")
print(f"Num: {num_children}")
print(f"Avg: {avg_age}")
````

# Unit 7: NumPy and the Big Problem with Lists

Lists are powerful datatypes in which we can store multiple types 

If they contain numbers we can perform relatively quick operations using built-in list methods and functions
- Min
- Max
- Sum
- Slicing
- Copy
- Append

<br>

## BUT....What if we want to do math easily and quickly on all elements of a list?
- Add an integer to every element
- Multiply by a value
- Add all the values of two lists together

<br>

#### Example: Multiply every item in a list by 7
>    How can we do this?<br><br>
>    Probably will need a loop

<br>

````python
# multiply 7 to every element in a list

somelist = [3, 5, 1, 2, 7, 88, 34, 24, 34, 121, 234, 6, 7, 8]
num = 7

print( somelist )

# iterate through each element
for i in range( len(somelist) ):
    somelist[i] = somelist[i] * num      # update the list
    
print(somelist)
    
````

<br>
<br>
<br>

### But...We can do this really quickly with a new library module called `NumPy`

<img src="https://i.postimg.cc/d1MrNqMm/07-numpy.png">


* Don't even need a loop


````python
# shorthand way of importing NumPy
import numpy as np


somelist = [3, 5, 1, 2, 7, 88, 34, 24, 34, 121, 234, 6, 7, 8]
numpyarray = np.array( somelist )  # make a numpy array from list somelist
num = 7

print( numpyarray )

# multiply the whole array by num
numpyarray = numpyarray * num    # or numpyarray *= num
print( numpyarray )
````

<br>

#### Like lists we can also find statistics BUT better:
- Min (and where the min occurs)
- Max (and where the max occurs)
- Sum
- Mean
- Standard Deviation
- etc

````python
print("NumPy statistics")
print('-' * 20)

print(f"Min: {np.min(numpyarray)} occurs at index {np.argmin(numpyarray)}" )
print(f"Max: {np.max(numpyarray)} occurs at index {np.argmax(numpyarray)}" )
print(f"Sum: {np.sum(numpyarray)}")
print(f"Mean: {np.mean(numpyarray):.2f}")
print(f"Std Dev: {np.std(numpyarray):.2f}")
````
