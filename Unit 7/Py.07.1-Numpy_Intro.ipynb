#### Let's first have a look at the `NumPy` array, and discuss how it differs from `lists`, `tuples`, and `strings`
  
  
<br>

#### A NumPy example

````python
import numpy as np    # import the NumPy library

# create a list, tuple, and array 
MyList = [ 45, 3.14, "hello" ]            # square brackets
MyTuple = ( 45, 3.14, "hello" )           # round brackets
MyArray = np.array( [45, 3.14, "hello"] ) # round & square  
                                          # np.array() function takes any sequence as an input
````

<br>

#### We can print the first elements of each sequence using **indices**

````python
print( "First element of list:", MyList[0] )
print( "First element of tuple:", MyTuple[0] ) 
print( "First element of array:", MyArray[0] )
````

<br>

#### Or the last element (or any inbetween)

````python
print( "Last element of list:", MyList[-1] )
print( "Last element of tuple:", MyTuple[-1] ) 
print( "Last element of array:", MyArray[-1] )
````

<br>

#### Or we can print the whole sequence and its `type`

````python
# print their contents
print( MyList )
print( MyTuple )
print( MyArray )

print()  # blank line

# investigate their type
print( type( MyList ) )
print( type( MyTuple ) )
print( type( MyArray ) )
````

<br>

#### At first glance, they don't look all that different!
* The NumPy array <u>does not</u> have commas separating the elements
* The elements in the NumPy array were all made into one datatype.  What type?

<br>
<br>

### Let's print the datatype of each element of the sequences

````python
# print the type of elements in each sequence

# the list
print( "*****THE LIST*****" )
print( MyList )
print()

print( "The List Content:" )
for i in range( len(MyList) ):
    print( "  ", MyList[i], type( MyList[i] ) )
````

````python

# the tuple
print( "*****THE TUPLE*****" )
print( MyTuple )
print()

print( "The Tuple Content:" )
for i in range( len(MyTuple) ):
    print( "  ", MyTuple[i], type( MyTuple[i] ) )
````

````python

# the ndarray
print( "*****THE NUMPY ARRAY*****" )
print( MyArray )
print()

print( "The ND Array Content:" )
for i in range( len(MyArray) ):
    print( "  ", MyArray[i], type( MyArray[i]) )
````

<br>

### What happened?
- The list and the tuple have what we'd expect: an integer, a float, and a string
-  The NumPy array *ONLY* contains strings (and they are "special" `numpy.str` types)
    - Not much different than "normal" strings<br><br>

* What is going on?
    - Arrays are **homogenous** data structures
    - **IMPORTANT: NumPy arrays can only contain data of the <u>EXACT SAME</u> type!**
    
    - If an array is created with different data types, NumPy will **"promote"** that data to the **most "complex"** type<br><br>

* #### NumPy Order of complexity:
    - strings > floats > ints

### Size and Shape of NumPy arrays
* We've seen that we can print the `type` of an array
* We can also find some useful information about our array:
    - `size`
    - `shape`
    - `number` of dimensions

````python
some2Dlist = [(1, 4, 6, 19), 
              (-34, 3.4, 4, 2), 
              (4, -67.3, 14, 900)]
              
NewArray = np.array( some2Dlist )  # this is a 2D array
print(NewArray)
print()

print( "length =", len( NewArray ) )
print( "size =", np.size(NewArray) )   
print( "shape =", np.shape(NewArray) )
print( "dimensions =", np.ndim(NewArray) )
````

> **Notice:** the `len` doesn't give us the number of elements.  More on this later

## Lists vs NumPy Arrays

<center><img src="https://i.postimg.cc/3N0NLvTH/07-01-lists-vs-numpy.jpg" width=400></center>

* So why use NumPy arrays instead of lists?

* **Recall:** when using a list, multiplying the list by a value simply made more <u>copies</u> of the list

<br>

### Example: High and low temperatures over a ten day forecast using Lists
* Calculate the average by adding the lists together and dividing by 2
* We will use the `random.sample` function to make a random list of highs and lows

````python
# make some lists
import random

highs = random.sample( range(35, 90), 10 )  # random sample of 10 highs
lows = random.sample( range(-10, 34), 10 )  # random sample of 10 lows

combined = highs +  lows   # add the lists together

print("highs:", highs)
print("lows:", lows)
print()

print("combo:", combined)
````

> This doesn't do what we want

<br>

##### If we really wanted to add the values of two lists together we would need a **loop** and a **new list**

````python
# blank list
combined_high_low = []

# iterate through the list
for i in range(len(highs)):
    added = highs[i] + lows[i]
    combined_high_low.append( added )
    
print("new combo:", combined_high_low)
````

<br>

##### What if we want to calculate the **average** of each day in the highs and lows list?

````python
combined_avg = []

for i in range( len(highs) ):
    avg = ( highs[i] + lows[i] ) / len(highs) 
    combined_avg.append( avg )

print("avg:", combined_avg)

````

> This is a lot of extra work

<br><br>

### Example: The power of the NumPy array
* What happens when we perform the same arithmatic on NumPy arrays?

````python
# imports
import numpy as np
import random

highs = random.sample( range(35, 90), 10 )  # random sample of 10 highs
lows = random.sample( range(-10, 34), 10 )  # random sample of 10 lows

# Convert these lists to NumPy arrays instead
highs_arr = np.array(highs)
lows_arr = np.array(lows)





# Now lets add them together
high_low_arr = highs_arr + lows_arr

print("highs:", highs_arr)
print("lows: ", lows_arr)
print()

print("combo:", high_low_arr)
````

> See how this added each element together?
> 
<br>

````python
# And compute our average temperatures (without using a loop!)

avgs_arr = high_low_arr / 2   # this divides every element by 2

print( "avg:", avgs_arr)
print( type(avgs_arr) )
print( type(avgs_arr[0]) )
````

## NumPy Broadcasting

#### Scalar operations applied to a NumPy array are **"broadcast"** to each element of the array.
```text
2 * [3 5 10] = [2*3  2*5  2*10] = [6 10 20]
```

<br>

#### With lists, multiplication replicates a list

````python
# with lists we replicate
mylist = [2, 4, 5, 7, 1, 2, 4, 5]

print("original:", mylist)
print("3 times: ", 3 * mylist)
````

<br>

#### With NumPy arrays, multiplication **broadcasts** to every element

````python
# with nd arrays we multiply
myarray = np.array(mylist)
print("original:", myarray)

mynewarray = 3 * myarray
print("3 times :", mynewarray)
````
