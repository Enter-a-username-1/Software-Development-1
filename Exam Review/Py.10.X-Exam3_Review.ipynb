## 3. Catching Exceptions

#### Use `try`/`except` to gracefully handle errors

#### The full block is 
```python
try:
    code that could give an error
except: 
    catch errors that may occur
else:
    do this when no errors occur
finally:
    do this whether there is an error or not
```

#### Catch any error and display the system error message

````python
try: 
    filename = 'myfile.txt'
    data = np.loadtxt( filename ) 
    
except IOError as e:
    print(f"an exception occurred {e}")
    print(f"exception type: {type(e).__name__}")
    
````

<br>

#### We can have multiple exceptions 
* add some more specific errors
* add the `else`

````python
try: 
    filename = 'data.txt'
    data = np.loadtxt( filename ) 

except FileNotFoundError:  
    print( "Cannot find", filename )
    
except Exception as e:
    print("an exception occurred:", type(e).__name__, e)
    
else:
    print( filename, "read successfully" )
    print( data )
````

<br>

#### One more try
* The `finally` displays no matter what

````python
try: 
    filename = 'data.txt'
    data = np.loadtxt( filename ) 

except FileNotFoundError:  # also IOError
    print( "Cannot find", filename )
    data = np.array( [] )

except ValueError:  # also IOError
    print( "Error reading", filename )

    # try reading again
    try: 
        print( "Trying different delimiter and header" )
        data = np.loadtxt( filename, delimiter=',', skiprows=1 ) 
    except Exception as e:
        print("an exception occurred", type(e), e)
        data = np.array( [] )
    
except Exception as e:
    print("an exception occurred", type(e), e)
    data = np.array( [] )
    
else:
    print( filename, "read successfully" )
    
finally:
    print( data )
````

## 4. Files
* In Python we can read and write files
* Always need to close files if we open them
* Best to use `with`/`as` operator to close the files automatically

<br>

### Files in Core Python
* `data = myfile.read()` reads the entire file as a string
* `data = myfile.readline()` reads one line as a string
* `data = myfile.readlines()` reads all lines into a list of strings



<br>

#### Read Files using `with`
* Use this!!!

````python
with open('text.txt', 'r') as myfile:
    for line in myfile:
        line = line.strip()
        print(line)
````

````python
with open('text.txt', 'r') as myfile:
    data = myfile.read()  # read all into a string
       
print(data)
````

<br>

#### Write Files

* `'w'` is for write
* `'a'` is for append

````python
file = 'newfile.txt'

with open( file, 'a') as myfile2:  # open the file newtext.txt to append to it
    myfile2.write("Behold the append, the kinder way to write\n")
    myfile2.write("It writes but does not\n")
    myfile2.write("write over what was previously written\n")

print( "Appending to", file, "complete" )
````

#### Plot the data
* plot x vs y
* put a horizontal line at the average y value
* include grid, labels, axes, legend, and title

````python
# create data
x = data[:, 0]
y = data[:, 1]

# find mean and make array of that value
avg_y = np.mean(y)
avg_line = np.full(np.size(x), avg_y)

# plot lines
plt.plot( x, y, label='x vs y' )
plt.plot( x, avg_line, label='avg y' )

# add labels
plt.title(' x vs y' )
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()

# limit x and y
plt.xlim( np.min(x), np.max(x) )


plt.show()
````
