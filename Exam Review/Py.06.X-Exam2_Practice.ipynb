### 1. Build a list of sines
Create a function called `create_sines`
>**Parameters**: starting and ending angles as integers<br>
>**Return**: list of sine values

#### Function Tasks:
1. Receive a starting and ending integer degree angle as parameters
2. Create a list of sines of all angles between the starting and ending angle in steps of 5 (inclusive) 
3. Return the list of sine values

#### Main code tasks:
1. In the main code, ask the user for a starting and ending angle
2. Pass these values to the function
3. Print the returned sins and the corresponding angles

#### Example output
```text
    sin of  0 is 0.000
    sin of  5 is 0.087
    sin of 10 is 0.174
    sin of 15 is 0.259
    sin of 20 is 0.342
```

````python
import math

# create_sines function
sines = []
def create_sines(start, end):
    for angle in range(start, end + 1):
        sine_value = math.sin(math.radians(angle))
        sines.append(sine_value)


# Main code

# get user angles
start_angle = int(input("Enter the starting angle: "))
end_angle = int(input("Enter the ending angle: "))

# create sines list from start to end
create_sines(start_angle, end_angle)

# print mysines with corresponding angle
for i, sine_value in enumerate(sines):
    angle = start_angle + i
    print(f"Angle: {angle}, Sine: {sine_value}")
````

### 2. Palindrome function
Create a function called `isPalindrome`
>**Parameters**: string<br>
>**Return**: None

#### Function Tasks:
1. Write a function that receives a string as a parameter
2. The function change the string to lower case
3. It then reverses the string and prints it
4. Print a message if the string and the reverse are equal or not (it is a palindrome)

#### Main code tasks:
1. Outside the function, ask the user for a string
2. Call your function with that string as an argument.

#### Example outputs
```text
    Enter string:  Racecar
    Reverse string is: racecar
    racecar is a palindrome

    Enter string:  puppy
    Reverse string is: yppup
    puppy is not a palindrome
```

````python
# isPalindrome function
def isPalindrome(word):
    word = word.lower()
    return word == word[::-1]


# main code
is_palindrome = input("Enter a word to check if it's a palindrome: ")
print("reverse string:", is_palindrome[::-1])
if isPalindrome(is_palindrome):
    print(f"{is_palindrome} is a palindrome.")
else:
    print(f"{is_palindrome} is not a palindrome.")

````
