#### With loops we need 3 things:
1. Initialize some value
2. Condition to check
3. Iterate our check value

#### Example: print the numbers 0-9

````python
# initialize variable i
i = 0        

while i < 10:
    print(i)
    i += 1    # iterate i by 1
````

<br>

### `while` loops are best with loops that we do not know how long they will take

#### Example: add values until -999 entered
* Note we can have an `else` on a loop

````python
count = 0

while True:
    var = eval( input( "Enter number to add: ") )

    if var != -999:
        count += var
    else:
        break

print( "The sum of all numbers is:", count )
    
````

<br>

### BEWARE: The Infinite Loop
* We loop forever **without** a way to exit
* Make sure you are iterating and have an exit condition

#### `for` loops do not have the same risk as `while` loops
* They are best for sequences
    * Numerical
    * Strings
    * Lists
    * Tuples

#### We have two types of `for` loop in Python
1. Walking through a sequence
2. Using the `range` function

<br> 

#### 1. `for` loops and sequences

````python
# Print all letters in "Sir Lancelot"
name = "Sir Lancelot"

for letter in name:
    print(letter, end = ' ')
````

````python
# Print a list of words
mylist = ['black knight', 'arthur', 'swallow', 'tim', 'spam', 'monty']

for word in mylist:
    print(word)
````

<br>

#### 2. `for` loops and the `range` function

<br>

##### `range( start, stop, step )`
* generates a sequence from `start` to `stop-1` by `some step`

<br>

##### `range()` can help with for loops

````python
# use a for loop to print the numbers 700-6000 by 500s

for i in range( 700, 6000, 500 ):
    print(i)
````

````python
# print spam 5 times

for i in range (5):
    print("spam")
````

<br>

#### Just like `while` loops, `for` loops can also have an `else`
* If we `break` the `else` will be skipped

````python
# print all values and their sum
mysum = 0

for num in range(20):
    print( num, end=", " )
    mysum += num
    
else:
    print( "Sum =", mysum )
    
````

### Random Module

We can get `random` functionality by using the `random` module
* `random.randint( low, high )` gives a random integer between low and high (inclusive)
* `random.random()` gives a random floating point value from 0-1 (inclusive)

#### Example: print a random integer from 0-10

````python
import random

print( random.randint( 0, 10 ) )
````

#### Example: print 20 random ints from 0-100

````python
import random

for i in range(1, 21):
    
    val = random.randint(0, 100) # from 0-100 inclusive
   
    print( i, val )
    
````

### Function Terminology in Python Code

- **define (`def`)**: the code that defines what the function will do
- **call**: use function in your program
- **pass**: how you transfer variables from the rest of your code to the function
- **argument**: the information you are passing to the function from your code
- **parameter**: the argument when it is used in your function (often used interchangeably with "argument")

### Two types of functions: Void and Valued

* #### **Void** Functions
    - Does some action
    - Does not `return` a value back
    - i.e. a function that prints a value out `print()`

<br>

* #### **Valued** Functions
    - Does some action
    - Will `return` a value to us
    - i.e.
        - `input()`
        - `math.cos()`

<br>

#### Example: a function that prints the volume of a cylinder
* void or valued?

````python
import math

def print_cyl_vol ( rad, h ):
    vol = math.pi * rad**2 * h
    return("Volume is", vol)
````

##### This function does nothing **until "called"**

````python
print_cyl_vol( 10, 5 )
````

<br>

### You **NEED** to do something with `return` values
* void or valued?


````python
def cyl_vol ( rad, h ):
    vol = math.pi * rad**2 * h
    return vol
````

````python
cyl_vol( 10, 5 )
print( "We forgot to use the return value??" )
````

````python
our_vol = cyl_vol( 10, 5 )
print( f"The volume is {our_vol:.2f}" )
````

<br>

### Most variables modified in a function are **NOT** available outside
* Functions are like Vegas!  What happens in the function stays in the function (with one exception)
* We only have access to returned (and saved) values

````python
print( vol )   # print vol from cyl_vol function
````

### Tuples

<center><img src="https://i.postimg.cc/CMBqG340/sequences-tuples.jpg" width=300></center>

##### VERY similar to lists
* They contain comma-separated values
* They use parentheses `( )` instead of square brackets `[ ]`
* They can even be made from `,` separated values
* They are **immutable**


````python
spam = ( "red", "blue", "green", 12345 )
print( spam )

spam = "red", "blue", "green", 12345
print( spam )
````

### Slicing Sequences

````python
mynewlist = [ 'a', 1, 217, 299, "monty", "python", 1999 ]
print( mynewlist )
````

````python
# change 4th item
mynewlist[3] = 46578 
print( mynewlist )
````

````python
# only the first 2 items
print( mynewlist[0:2] )
````

````python
# only the last 2 items
print( mynewlist[5:7] )
````

<br>

#### Can change multiple values in a list with a slice
* The slice size does not have to match the replacement value

````python
print( mynewlist )

mynewlist[1] = [-99]

print( mynewlist )

mynewlist[2] = [ 'spam', 'eggs', 'bacon', 'spam' ]

print( mynewlist )
````

### Sequence Methods
* Methods are a special function that is part of the datatype
* Functions we call by name and pass the variable in the parentheses: `print(somestring)`
* Methods use "dot notation" after the variable name: `somestring.count()`

<br>

#### String methods

````python
my_string = "the first lesson in Not Being Seen - not to stand up"

print( "Formatting" )
print( "----------------------" )
print( "  Original   ",my_string)
print( "  lower()    ",my_string.lower() )             
print( "  upper()    ",my_string.upper() )             
print( "  capitlize()",my_string.capitalize() )    
print( "  title()    ",my_string.title() )             
print( "  swapcase() ",my_string.swapcase() )    

print()

print( "Find and manipulate")
print( "----------------------")
print( "  find('r')          ",my_string.find("r") )            
print( "  replace('in','##') ",my_string.replace("in","##") )
print( "  count('i')         ",my_string.count("i") )          
print( "  split()            ",my_string.split() )   # splits on space
````

````python
my_string = '59'

print( "Checking", my_string)
print( "----------------------")
print( "  isdecimal()   ",my_string.isdecimal() )     
print( "  isalum()      ",my_string.isalnum() )          
print( "  isalpha()     ",my_string.isalpha() )         
print( "  isdigit()     ",my_string.isdigit() )   
````

#### List methods & functions

````python
mylist = [ 'spam', 'eggs', 'bacon', 'spam', 'and', 'sausage' ]
print( mylist )
````

> ##### `len()`

````python
print( len( mylist ) )
````

> ##### `.append()`

````python
mylist.append( 'baked beans' )
print( mylist )
````

> ##### `.index()`

````python
print(f"Bacon is found at index {mylist.index('bacon')}")
````

> ##### `del` function by index

````python
del mylist[4]  # delete 'and'
print( mylist )
````

> ##### `.remove()` by value

````python
mylist.remove( 'sausage' )
print( mylist )
````

> ##### `.remove()` all values

````python
to_remove = 'spam'

for item in mylist:
    if item == to_remove:
        mylist.remove(item)

print( mylist )
````

> ##### `.sort()`

````python
print( mylist )
mylist.sort()
print( mylist )
````

> ##### `sorted()`

````python
mylist = [ 'spam', 'eggs', 'bacon', 'spam', 'and', 'sausage' ]
mylist2 = sorted(mylist)

print( mylist )
print( mylist2 )
print( mylist )
````
