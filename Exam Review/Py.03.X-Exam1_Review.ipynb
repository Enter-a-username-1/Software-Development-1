## Operators & Expressions
#### We can use operators to perform basic math expressions
Our operators are:
* `+`
* `-`
* `*`
* `/` - floating division
* `//` - integer division
* `%` - modules aka remainder
* `**`

````python
# some basic math operations
var1 = 7
var2 = 4

print("add         ", var1, "+", var2, " =", var1 + var2)
print("subtract    ", var1, "-", var2, " =", var1 - var2)
print("multiply    ", var1, "*", var2, " =", var1 * var2)
print("divide      ", var1, "/", var2, " =", var1 / var2)
print("floor divide", var1, "//", var2, "=", var1 // var2)
print("modulus     ", var1, "%", var2, " =", var1 % var2)
print("exponent    ", var1, "**", var2, "=", var1 ** var2)
````

#### We can nicely format our output using the `f-string` method and `{}` brackets

<br>

### Rounding with f-strings
#### We can round our numbers to a specified significant figure
* put the number of decimals in the { } brackets
* use a colon ( : ) a period ( . ) and the number followed by f
```text
    {:.5f} for 5 decimal place float
```

````python
print( f"Your height is {height:.2f}in" ) # 2 decimals
````

<br>

### Commas to separate thousands places
```text
    {:,}
```

````python
num = 64040574382306.678043

print( f"my number is {num:,.2f}" )
````

<br>

### Columns
#### The column width can also be changed to make the output line up nicely
* put the width of the column in the `{ }` brackets after the colon ` :`
```text
    {:4} for 4 character width
    {:<4} left align and have 4 characters wide
```

````python
name1 = "Arthur"
name2 = "Tim"
name3 = "Black Beast"
age1 = 42
age2 = 275
age3 = 8045

# print a table
print( f"{'name':15} {'age':4}" )
print( "--------------------------------")
print( f"{name1:15} {age1:<10}" )
print( f"{name2:15} {age2:<10}" )
print( f"{name3:15} {age3:<10}" )
````

## Math Functions

#### Python has built-in math functionality
* Besides these operators: `+, -, *, /, //, %, **`

<br>

#### Python can add complex math functionality by using the `math` library
* Need to `import math`
* To use, call with "dot notation" ie `math.pi`
* We can use `math.sin`, `math.sqrt`, `math.hypot` and many more
* The trig functions require that the angle be in <u>radians</u>

##### Sin example

````python
import math

PI = math.pi  # this is a constant so ALL CAPS

sin = math.sin(PI / 2)
print( "sin PI/2:", sin )

rad_angle = math.radians(60)
print( f"60 degrees is {rad_angle:.2f} radians" )
````

##### Square Root Example

````python
squareroot = math.sqrt(2)

print( "square root of 2:", squareroot )
print( f"round to 2 decimals: {squareroot:.2f}" )
````

## Flow Control/Branching
#### Changing the flow of a program based on True/False questions makes programs really powerful
```Python
if condition == True:
    # do this thing
elif condition == True:
    # do this other thing
else:
    # do this if none of the above are true
```

<br>

##### Example: Get the day of the month

````python
# get the day of the month
date = eval( input( "What is today's date? " ) )

# check to see how far through the month we are
if date <= 5:
    print( "The month is just beginning" )
elif 5 < date < 25:
    print( "You are in the middle of the month" )    
elif date >= 25:
    print( "the month is almost over" )
````

<br>

#### We can combine logical and conditional operators 

##### Example: Check for Christmas

````python
# find the month and date
date = eval( input( "What is today's date? " ) )
month = eval( input( "What number month is it? " ) )

if date == 25 and month == 12:
    print( "Merry Christmas" )
else:
    print( "Not Christmas yet" )
````
