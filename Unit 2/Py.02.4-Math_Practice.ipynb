> Recall that the pythagorean theorem is: $a^2 + b^2 = c^2$

* ask the user for a and b 
* calculate c and print it out to 2 decimal places

<br>

**Example input/output:**

```text
    Enter a:  3
    Enter b:  4

```

**Another Example input/output:**

```text

```

### 4. Future value of savings/investment
* Compound interest can be computed with the following formula

    $FV = Pe^{rt}$

    Where:
    - $FV$ = future value
    - $P$ = initial principal
    - $e$ = Euler's number
    - $r$ = rate (decimal)
    - $t$ = time in years
 
 * Ask the user to enter a starting deposit, the interest rate as a decimal value, and the number of years. Calculate and display the final value.
 
* Your result should look like:

    ```text
    Starting amount:  1000
    Interest rate:  .02
    Number of years:  14

    Your $1000 will grow to $1323.13
    ```


````python
# get inputs
initial_investment = float(input("Enter the initial investment amount: "))
interest_rate = float(input("Enter the annual interest rate: "))
years = int(input("Enter the number of years the money is invested: "))
# calculate future value
future_value = initial_investment * math.e**(years * interest_rate)

# display result
print(f"The future value of your investment of ${initial_investment:.2f} is: ${future_value:.2f}")

````

### 6. Log Functions
* Recall that the log function can tell us how many numbers need to be multiplied together to get a result
* Ex: how many 5s need to multiplied together to get 625?
    - $ 5 \cdot 5 \cdot 5 \cdot 5 = 625 $ we need 4 of the 5s to get 625 or
    - $ log_5(625) = 4 $<br><br>
    
* **Ask the user for a number $x$ and a base $ y$ and find the $log_y(x)$**

* Your result should look like:
    ```text
    
    Number to take log of:  625
    Base of log:  5
    The log base 5 of 625 is 4.0
    ```

````python
# ask for input
angle = float(input("Enter a number to calculate its log: "))
base = float(input("Enter a log base: "))
# calculate log
log_value = math.log(angle, base)

# display output
print(f"The log base {base} of {angle} is: {log_value}")
````
