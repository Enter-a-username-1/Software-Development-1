# String Formatting

Python has several methods for formatting output (%, format, f-string).  We will focus on the most recent method - **The F-String**.

## The F-String

#### The print function will output a string to the screen.
- It "casts" whatever is between the parenthese as a string
- We can format this output in many different ways

### Format-Strings in their most basic form:

```python
print( f"message {variable0} {variable1}" )
```

### Lets try out `f-strings`

Let's calculate the area of a rectangle

````python
# Compute the area of a rectangle

l = 200.7261
w = 50.001
a = l * w

print("The area of a rectangle with length", l, "and width", w, "is", a, "ft^2")
````

<br>

#### What if we store the units as a variable?

````python
# Compute the area of a rectangle

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print("The area of a rectangle with length", l, "and width", w, "is", a, units )
````

<br>

#### We can print these variables using `f-strings` instead
`f-string` replaces the variable name inside `{ }` with its value

<br>

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( f"The area of a rectangle with length {l} and width {w} is {a} {units}" )
````

> Okay, but we don't need .5057261 on our area.  <br>Let's round it

<br><br>

### F-String Rounding: 
#### Lets display our dimensions to <u>1 decimal</u> and results to <u>2 decimals</u>
* In the `{ }` we can place a `:`
* we then put a `.` , a `number` and `f` (for float)
    * example `{variable:.2f}` will give us a variable with 2 floating point decimals

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( f"The area of a rectangle with length {l:.2f} and width {w:.2f} is {a:.2f} {units}" )
````

> That's better.  Note that the `f-string` rounds the number displayed (not just truncating it).
> <br><br>Maybe we want to separate the thousands with commas

<br>

### Formatting: Commas at thousands
* We can change `:.2f` to `:,.2f`

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( f"The area of a rectangle with length {l:.1f} and width {w:.1f} is {a:,.2f} {units}" )
````

<br><br>

### Formatting: Display aligned as a table
* Add a number after the `:` to "reserve" that many spaces
* Can also use `>` or `<` to align the information

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( "Rectangle Data" )
print( "----------------" )
print( f"{"length"}{"width"}{"area"}" )
print( f"{l:.1f}{w:.1f}{a:,.2f}{units}" )   
````

> What a mess!!!!  Let's fix that by setting column sizes

<br>

### Formatting Column sizes

* Let's reserve 10 spaces for the length and width columns
* And 15 spaces for area column

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( "Rectangle Data" )
print( "----------------" )
print( f"{"length":10}{"width":10}{"area":15}" )
print( f"{l:.1f}{w:10.1f}{a:15,.2f}{units}" )  
````

> A little better, but the words are not lined up with the numbers

<br>

### Formatting Alignment
* Numbers are aligned <u>right</u> by default
* Words are aligned <u>left</u> by default
> Just like Excel

* We can use 
    - `>` to right align
    - `<` to left align
    - `^` to center

````python
# Compute the area of a rectangle - print with format

l = 200.7261
w = 50.001
a = l * w
units = "ft^2"

print( "Rectangle Data" )
print( "----------------" )
print( f"{"length":<10}{"width":^10}{"area":>15}" )
print( f"{l:10.1f}{w:10.1f}{a:15,.2f}{units:>5}" )  
````

<br><br>

#### Note that `f-strings` round to the desired decimal place - it doesn't truncate like type casting does!

* There a numerous other "options" for formatting!
    - alignment
    - comma separation
    - scientific formatting
    - percentages
    - hex, octal and binary

### Formatting Strings Directly
#### Also note that we don't have to apply the `f-string` only inside of a `print` statement, it will work on a string itself
### Formatting Strings Directly
> We use `MYPI` in all caps because PI is a constant

````python
MYPI = 3.14159265358979

MyStr = f"Full PI {MYPI}, but it's best to only use a little bit of PI {MYPI:.4f}"
print(MyStr)
````

The placeholder `{:.4f}` limits pi's output to just 4 decimal places

````python
str1 = "Spam" 
str2 = "eggs"
str3 = "sausage"

composite = f'{str1}, {str2}, {str3} {"& spam"}'
print(composite)
````

## Some other formatting information
 - **Escape codes** can be used to further control output
     - `\n` is the **newline**, it causes what follows to be on the next output line
     - `\t` is the **tab**, causing output to move to the next multiple of 8 column (don't use this much)  
<br><br>

 - `print( )` normally adds a **newline** when it completes. This can be overridden with the `end="something"` parameter
     - `print( "Stay on this line", end='' )` # nothing between the single quotes
     - `print( "Double space now", end='\n\n' )`  # two newlines will be executed, leaving blank line before next print
     - `print( "Ending sentences automatically", end='.\n' )`  #period and newline when print executes
<br><br>

 - Similarly, print puts a **space** between comma separated items. This is overridden with `sep="something"` parameter
     - `print( var1, var2, var3, sep='' )`  #puts the output values right up against each other
     - `print( var1, var2, var3, sep='--' )`  #puts two dashes between the output values
     
<br>
     

#### Example: Printing multiple newlines

````python
print("This prints normally")
print("This prints with one blank line below", end="\n\n")
print("This adds two blank lines below", end="\n\n\n")
print("Normal print again")
````

<br>

#### Example: Printing on the same line by changing `end=`

````python
# printing on the same line
print( "I am printing on one line. ", end='' )
print( "This is on the same line too." )
print( "This is on a new line." )

print()

print( "This line will have two lines after it.", end='\n\n' )
print( "This line is two lines down." )
print( "And the spacing is back to normal" )
````

<br>

#### Example: Printing separated by elipses using `sep=`

- Similarly, `print` puts a **space** between comma separated items. This is overridden with `sep="something"` parameter
     - `print( var1, var2, var3, sep='' )`  #puts the output values right up against each other
     - `print( var1, var2, var3, sep='--' )`  #puts two dashes between the output values

````python
# print separated by elipses 
print( 1, 2, 3, 4, sep='...' )
````

````python
num1 = 15
num2 = 233
num3 = -56

# print numbers separated by commas
print( "Separated by commas" )
print( num1, num2, num3, sep=',' )

# print numbers separated by tabs
print( "Separated by tabs" )
print( num1, num2, num3, sep='\t' )
````

### Binary, Hex or Scientific
Sometimes we want to display numbers in other forms or bases
* We typically count in "base 10"
* Integers can be shown in base 2, 8 or 16 representations
* Floating point numbers can be displayed in scientific notation


````python

NumConv1 = ( f"We can display our numbers in many different forms\n" 
             f"20 to base 2 ({20:b})\n"
             f"25 to base 8 ({25:o})\n"
             f"30 to base 16 ({30:x})\n"             
             f"or scientific notation ({58674635374597868.009876:e})" )

print(NumConv1)
````
