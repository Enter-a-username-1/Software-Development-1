### 1. Cost of Swallow Eggs
* Print a line like:  5 swallow eggs costs $nnnnnnnn.nnnnnn.  
* Calculate the cost in the print statemment. 
    > Do not use any f-strings - just use commas.

````python
num_eggs = 12
egg_cost = .23
dozen_eggs = num_eggs * egg_cost

# print here - don't use f-string yet
print("A dozen eggs cost ", round(dozen_eggs, 2))
````
