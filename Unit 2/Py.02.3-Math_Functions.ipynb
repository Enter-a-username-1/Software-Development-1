## Built-In Math

#### Python has a number of built-in arithmetic operators, some we are very familiar with

#### Absolute Value

````python
# Absolute value
print( "Absolute Values" )
print( "absolute value of -5 =", abs(-5) )
print( "absolute value of -23.45 =", abs(-23.45) )
print()  # makes a blank line
````

#### Rounding
* `f-string` <u>only</u> changes how the value is displayed


````python
# Rounding f-strings
print( "---- Rounding with f-string ----" )

num = 234.567
print( f"Round {num} to a 2 decimals = {num:.2f}" )
print( f"Round {num} to a whole number = {num:.0f}")

print( "num is still", num)
````

* `round` <u>permanently</u> modifies the value

````python
# Rounding permanently
print( "---- Rounding with round function ----")

num = round(num, 2)
print( "Round 234.567 to a 2 decimals =", num )
print( "num is now", num)

num = round(num)  # round to 0 decimal places
print( "Round 234.567 to a whole number =", num )
print( "num is now", num)
````

#### Power
* Alternative to `**`
* Does the same thing.  Has optional 3rd argument for modulus

````python
# Power - can also be done with **
print( "Exponents" )
print( "2^8 =", pow(2,8) )
print( "5^3 =", pow(5,3) )
print( "5^3 % 7 =", pow(5,3,7) )
````

### Number related `math` functions
* `math.floor(x)` - returns integer below parameter
* `math.ceil(x)` - returns integer above parameter
* `math.fabs(x)` - returns absolute value of parameter (always as float)
* `abs(x)` - built-in function that will return int or float, depending on parameter type



````python
import math

high = 89.43
low = -18.4
intlow = -23

print( "floor of", high, "=", math.floor(high) )
print( "ceiling of", high, "=", math.ceil(high) )
print( "floating abs of", high, "=", math.fabs(high) )
print( "abs value of", high, "=", abs(high) )
````

````python
print( "floor of", low, "=", math.floor(low) )
print( "ceiling of", low, "=", math.ceil(low) )
print( "floating abs of", low, "=", math.fabs(low) )
print( "abs value of", low, "=", abs(low) )
````

````python
print( "Difference between fabs and abs" )
print( "floating abs of", intlow, "=", math.fabs( intlow ) )
print( "abs value of", intlow, "=", abs( intlow ) )
````

### Exponents and logs
* `math.pow(x,y)` - raises x to the power y. (Use ** for integer base and exponents)
* `math.sqrt(x)` - returns square root of parameter
* `math.exp(x)` - raises *e* to the power x. Better than math.e**x or pow(math.e, x )

<br>

* `math.log(x)` - natural log of parameter
* `math.log(x,y)` - log base y of x
* `math.log10(x)` - base 10 log of parameter
* `math.log2(x)` - base 2 log of parameter


````python
# find fractional roots
print( "raise 13^2.5", math.pow( 13, 5/2 ))  # could also use exponent 2.5

# another way to do it
root = 13 ** 5  # raise 13 to 5th power
root = math.sqrt( root ) # take square root
print( "another way 13^2.5", root )

# formatted nicely to 5 decimals
print( f"better format {root:.5f}")

print("\n") # prints two new lines
````

#### Log Example: Binary search
* The max steps in a binary search

````python
# find maximum number of steps in binary search 
# a method that reduces search area by half each step
# defined as the log base 2 of the data set

data_size = int( input( "How many data items for a binary search: " ))

steps = math.ceil( math.log( data_size ))

print( "Binary search will take at most", steps, "steps" )
````

### Trig functions
* All the common trig functions are available, as are their inverses
* The basic functions take angles in **radians**, the arc-functions return radian values
    - `math.sin(x)`
    - `math.cos(x)`
    - `math.tan(x)`    
<br>
    - `math.asin(x)`
    - `math.acos(x)`
    - `math.atan(x)`
    
    > ##### NOTE: All trig functions <u>require</u> the angle to be in **radians**

<br>
   
    
* There are also corresponding hyperbolic functions, add an 'h' to the end of the function names
    - `math.sinh(x)`
    - etc
<br>
<br>


* Python provides functions for converting degrees and radians
    - `math.radians(x)`  - converts degrees to radians
    - `math.degrees(x)`  - converts radians to degrees
<br>
<br>

* Distance functions (examples below in two dimensional plane, multiple dimensions are supported)
    - `math.hypot(x,y)`  - returns distance from origin to coordinate
    - `math.dist( (x1,y1), (x2,y2) )`  - returns distance between two points

<br>
    

#### Example: Find the height of a Tree
<center><img src="https://4.bp.blogspot.com/-v6rArJmw33k/TkFOlxikauI/AAAAAAAACZQ/AwQOp_zO--U/s1600/tp2ch1_image3.jpg">
<br>
&nbsp;&nbsp; <font size=1>Source: https://malini-math.blogspot.com/2011/08/applications-of-trigonometry-in-real.html?m=1</font></center>

````python
# find height of tree

# get input
angle = eval( input( "Enter angle θ from ground to top of tree (degrees): " ))
dist = eval( input( "Enter distance x from tree (ft): " ))

# convert to radians
angle_rad = math.radians( angle ) 

 # use tangent to calculate height
height = dist * math.tan( angle_rad ) # tangent

print()  # extra line

print( f"The tree is {height:.2f} ft high" ) # 2 decimals

string_len = math.hypot( dist, height )

print( f"You need a string at least {string_len:.2f} ft long to go from the ground to top of tree" )
````
