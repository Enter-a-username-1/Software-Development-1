## Recap: Iteration

#### Last time we learned:
- How to repeatedly execute a block of instructions with `while`
- The dangers of the dreaded <font color=red>**infinite loop**</font>
- How to break out of a loop with `break`

<br>

### Iteration: there are <u>always</u> three aspects
1. Initialize loop variable
2. Loop condition
3. Iterate the loop variable

We did this with the `while` loop 

<br>

#### Example: `while` loop, is it raining?
<img src="https://i.postimg.cc/ZRL6FHYW/is-raining-loop.png" width=400>

````python
# while loop of is it raining?

raining = bool(input(("Is it raining? (True/False)")))                # intialize variable

while ( raining ):            # loop condition same as raining == True
    
    umbrella = input("Do you have an umbrella? (yes/no)")
    
    if umbrella == "yes":
        break                 # break out of the loop
    else:
        print("Wait a while")
        
        
    
    stillrain = input("is it still raining? (yes/no)")   # string input from user
    
    if (stillrain == "yes"):
        raining = True        # change loop variable
    else:
        raining = False       # change loop variable
        
        
print("go outside")
````

# The `range()` function

Before we get into our next loop structure, we need to talk about Python's `range` function

* The `range` function generates a list of numbers between a **start** and **stop**.  
    * The default is to start at 0
    * The stop is <u>exclusive</u>
* The list will change with some **step** value
    * The default step is 1
    
    ```python
    range( start, stop, step )
    ```


<br>

#### Example: Range of 0-10 by 1s

````python
# use range to generate numbers 0-10 using a step of 1

myrange = range(0, 10, 1)
print(myrange)
````

````python
type(myrange)
````

> Well that is not what we expected

<br>

#### Let's change that to a `list`
> A list is a special Python **datatype** we will look at more in Unit 6

````python
# use range to generate a list of numbers 0-10 using a step of 1

myrange = range(0, 10, 1)
print( list(myrange) )    # typecast to a list to show the numbers
````

> That is only **0-9**.  We wanted **0-10**

````python
# use range to generate a list of numbers 0-10 using a step of 1

myrange = range(0, 11, 1)   # (start, stop, step)
print( list(myrange) )
````


#### **Notes:**
1. The only required input to `range()` is the stop value
2. The "range" of values is really `start` to `stop-1`
3. The start, stop, step **MUST** be <u>integers</u><br><br>

* You can futher familiarize yourself with the `range()` function <a href="https://www.w3schools.com/python/ref_func_range.asp">HERE!</a>!


<br>

#### If only one value is in `range()` it is assumed to be the stop
* Unless specified assume:
    * `start = 0`
    * `step = 1`

````python
# implied start and step values

myrange = range(11)        # 11 is the stop. 
print( list(myrange) )
````

<br>

#### If two values are in `range()` they are the start and stop
* Assume `step = 1`

````python
# implied step value
myrange = range(0, 1)    # 1 is the step
print( list(myrange) )
````

<br>

#### We can also modify the step values

````python
# 0-10 by twos
myrange = range(0, 11, 2)  
print( list(myrange) )
````

````python
# 0-10 by fours
myrange = range(0, 11, 4)
print( list(myrange) )
````

````python
# 10-0 in reverse
myrange = range(10, -1, -1)   # start at 10 and go to 0
print( list(myrange) )
````

````python
# 10-0 in reverse by twos
myrange = range(10, -1, -2)
print( list(myrange) )
````

#### So, you see that `range` can quickly generate a list of numbers.  This is very helpful for iteration

# The <code>for()</code> loop

### There are essentially two different ways to iterate in python
- We talked about the `while` loop

### `for` loops are counting loops
* They are "safer" than `while` loops
* They are clearer than `while` loops
* They require fewer lines of code than `while` loops

<br>

### When dealing with the `for` loop, we use two methods for iterating
1. `range()`
2. a sequence like a `list` or `string`

<br>

<center>
    
### Basic syntax of a `for` loop

### We will first look at using `range()` with `for` to control the iterations
 
```python
    for i in range(start,stop,step):
        # Block of code to execute if i doesn't meet the stop condition
```

* You can futher familiarize yourself with python for loops <a href="https://www.w3schools.com/python/python_for_loops.asp">HERE!</a>!


<br>

#### Let's first see how the `range()` function works with a `for` loop 

````python
# range() function with a for loop 
n = 10

for i in range(n):                # assume start=? step=?
    print("The value for i =", i)
    
print("We've now left the for loop")
````

> **Note that:**<br>
    1. The loop was executed $n = 10$ times<br>
    2. The loop counter $i$ is incremented as the loop executes<br>
    3. When $i == (n-1)$, the loop terminates<br>
    4. The default start value is 0

#### We can also add the start and step conditions as variables to have more control

````python
# change start and step conditions to have more control 
start = 0
stop = 10
step = 2

for i in range( start, stop, step ):
    print("The value for i =",i)

print("We've now left the for loop")
````

<br>

#### We don't need to use $i$ in the loop.  We can print something else

````python
# print swallow 10 times

for x in range (10):
    print("swallow")
````

<br>

#### With `range` we can iterate large numbers easily

##### Example: Numbers 0-6000 by 500s inclusive

````python
# use a for loop to print the numbers 0-6000 by 500s

for i in range( 0, 6001, 500 ):
    print(i)
````

> **NOTE:** Start/stop/step <u>**must**</u> be integers

````python
# What happens if the start/stop/step conditions aren't integers? 
incr = 0.2

for i in range( 0, 5, incr):
    print(i)
````

<br>

### Where are our three parts of a loop?
1. initialize
2. condition
3. iterate

````python
# 0-600 by 50s again

for i in range( 0, 601, 50 ):   # intialize/condition/interate?
    print(i)
````

## We can also use `for` loops to iterate through a sequence
* One type of sequence is a `list` of numbers, values etc
* Another type of sequence is a `string`, which is really a list of letters
* We will learn more about sequences in **Unit 6**

<br>

### `for` loop and a list of numbers
* Python lists are designated by `[ ]`
* Values are separated by commas
* We use the `in` operator to loop through each item in the list with a `for` loop

````python
mylist = [3, 6, 12, 14, 99, 103, 23, 42]  # list of numbers

for num in mylist:
    print(f"{num} squared is {num**2}" )
````

<br>

### `for` loop and a list of words

````python
mylist = ['black knight', 'arthur', 'swallow', 'tim', 'spam', 'monty']

for word in mylist:
    print(word)
````

<br>

### `For` loop and a sequence of characters AKA **a `string`**
* Strings are <u>just</u> arrays or sequences of characters
* Each character can be accessed with an index
* Indices start at 0

#### Example: Print all letters in "Sir Lancelot"

````python
# Print all letters in "Sir Lancelot"
name = "Sir Lancelot"

for letter in name:
    print(letter, end=' ')  # recall that we can change the 'end' of a print
````

## `continue`

<center>
<img src="https://i.postimg.cc/c4BT7v3r/04-python-for-loop-with-continue-statement.png" width=500>
<figcaption>https://codebuns.com/python/python-for-loop/</figcaption>
</center>


### The `continue` statement jumps to the next iteration
* It is the opposite of `break`
* Sometimes we want to skip over values
* `continue` stops the code below it, and goes to the next iteration step


#### Example: Print all letters in "Sir Lancelot" but ignore vowels

````python
# Print all letters in "Sir Lancelot" but ignore vowels
name = "Sir Lancelot"

for letter in name:
    if letter == "i" or letter == "a" or letter == "e" or letter == "o":  # any vowels
        continue  # skip over print
        
    print(letter)
````

## Nested Loops
* Just as you could nest `if` statements as the action of `if..elif..else` blocks, you can nest loops within loops.
* This can make the code confusing.  Use with **caution**

<br>

<center>
    
### Syntax of nested `for` loops
<img src="https://i.postimg.cc/tgY5HDMx/04-python-nested-for-loop.jpg" width=500>
<figcaption><font size=1>https://codebuns.com/python/python-for-loop/</font></figcaption>
</center>


````python
# print the inner and outer parts of a loop
outer = 3
inner = 4

for i in range(outer):    
    print("outer", i)
    
    for j in range(inner):        
        print("  inner", j)
        
````
