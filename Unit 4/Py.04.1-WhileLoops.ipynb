## Repetition and Iteration

<br>

### What is repetition?
* It is doing the **same** thing again and again

### What is iteration?
* It is doing the same thing over and over by **changing** the value of **one** (or more) variables

<br><br>

#### For example, what if we want our program to count to 10? 
* It is pretty simple to do this

````python
print("x = 1" )
print("x = 2" )
print("x = 3" )
print("x = 4" )
print("x = 5" )
print("x = 6" )
print("x = 7" )
print("x = 8" )
print("x = 9" )
print("x = 10" )
````

### To use a `break` we:
* Make a loop that never ends
* Make a way to force us out of the loop using the `break` command

````python
# To use break we construct a loop that never ends 

while True: # This is now always true
    name = input("Please enter your name:")
    
    if (name.lower() == "you"):    # if name is "you" we end the loop
        break
        
print("Whew! That was a close one.")

````

## Inclusive vs Exclusive
* Whether the numbers are included or excluded in the loop
    * **inclusive** = use the end points
    * **exclusive** = ignore the end

````python
# 0-10 inclusive

num = 0            # initialize

while num <= 10:   # include 10
    print(num)
    
    num += 1       # iterate
````

````python
# 0-10 exclusive

num = 0           # initialize

while num < 10:   # exclude #10
    print(num)
    
    num += 1      # iterate
````

## Using `else` with a `while` loop
* While loops can take a single `else` condition like the `if`

<center>
    
### Syntax of a `while` loop with an `else`
<img src="https://i.postimg.cc/kg47ydkd/04-python-while-loop-with-else-clause.jpg" width=500>
<figcaption>https://codebuns.com/python/python-while-loop/</figcaption>
</center>

* This will run **once and only once** after the while condition is no longer `True`



````python
num = 0

while num < 10:
    print(num)
    num += 1
else:
    print("end of loop")
````


> **NOTE** `else` will not run if we exit loop with a `break`

````python
num = 0

while num < 10:
    print(num)
    num += 1

    if num % 7 == 0:
        break   # break when the number is evenly divisible by 7

else:
    print("end of loop")

print("We can now move on in our code")
````

#### Example: nested `while` loops for printing a pattern

````python
# print a pattern
end = 5

print("Pattern Printing using Nested While loop")
  
i = 0
  
while i <= end:
    print("")
    
    j = 0
    
    while j < i:
        print(j, end = ' ')
        
        j += 1
    
    i += 1
````

#### Example: print the multiplication grid
* <font color="green">Don't worry about understanding all this code</font>
    * It is here to show one possible use case
    * Most of this code is aligning columns


````python
# initialize variables
row = 1
col = 1
size = 10    # 10x10 multiplication table

# print header
print(f"{'Multiplication Table':^60}")


print( f"{' ':3}", end='' )    # prints a blank space in left column

while col <= size:
    print( f"{col:>5}", end='' )  # print number column headers
    col += 1
    
print()                               # blank line
print( "-" * 60 )                     # header line

col = 1                               # reset col  back to 1

# print mulitiplication values

while row <= size:
    print( f"{row:^3} |", end='' )    # print row numbers in far left column
    
    while col <= size:
        print( f"{col*row:^5}", end='' )
        
        col += 1                             # iterate col 
        
    print()
    
    col = 1                                  # reset col back to 1
    row += 1                                 # iterate row     
````

### `break` and scope
* **scope** is how long a command (like `break`) can affect our program
* These only affect the loop level in which they reside. 
* If inside a nested loop, they only exit or resume the interior loop.

````python
while True:
    
    num = int(input(">"))   # display input prompt
    
    while num < 5:
        print( num )
        num += 1
        
        if num == 3:
            print("Inner break when num == 3")
            break     # this only breaks the inner while
  
    if num > 10:
        print("Outer break when num > 10")
        break   # this breaks the main while loop
             
````
