### Comments on Code Formatting
Code is **read** more often than **written**

##### Goal: Make your code readable and consistent
* Fix the following code with:
    * variable names
    * spacing
    * comments

````python
import math
var1=float(input("Enter beacon horizontal offset (km): "))
var2=float(input("Enter beacon vertical offset (km): "))
if var1==0:
    print("CK: Beacon signal too unstable for angle calculation")
    VAR3=-99999
    var4="undefined"
else:
    VAR3=math.degrees(math.atan(var2/var1))
    print(f"Drift Angle: {VAR3:.3f} degrees")
    if VAR3<10:
        print("CK: Beacon drift within acceptable limits.\n")
        var4="acceptable"
    elif VAR3<25:
        print("CK: Warning: Beacon drift elevated.  Monitor closely.\n")
        var4="elevated"
    else:
        print("CK: Drift exceeds safety thresholds.  Recommend shutdown.\n")
        var4="excessive"
print(f"Drift angle is {VAR3} with a status of {var4}")
````

````python
#imports
import math

#inputs for data
horizontaloff=float(input("Enter beacon horizontal offset (km): "))
verticaloff=float(input("Enter beacon vertical offset (km): "))

#Calculations and output
if horizontaloff==0:
    #Runs if data does not allow for angle calculation
    print("CK: Beacon signal too unstable for angle calculation")
    beaconAngle=-99999
    beaconStatus="undefined"
else:
    #Runs if data allows for angle calculation and solves for angle and status
    beaconAngle=math.degrees(math.atan(verticaloff/horizontaloff))
    print(f"Drift Angle: {beaconAngle:.3f} degrees")
    if beaconAngle<10:
        print("CK: Beacon drift within acceptable limits.\n")
        beaconStatus="acceptable"
    elif beaconAngle<25:
        print("CK: Warning: Beacon drift elevated.  Monitor closely.\n")
        beaconStatus="elevated"
    else:
        print("CK: Drift exceeds safety thresholds.  Recommend shutdown.\n")
        beaconStatus="excessive"
#Output of final angle and status
print(f"Drift angle is {beaconAngle} with a status of {beaconStatus}")
````

### Good code formatting

* place **blank lines** between logical sections

* Good variable names

* <u>Only</u> constants are ALL-CAPS
* Keep column width under 100 characters
* Comments are an art
    - You must have them
    - Too many can make the code unreadable
    - Rarely use comments on the right

* Space out operators
    ```python
    # good
    i = i + 1

    # not good:
    i=i+1
    ```

#### Or what if we wanted to keep printing an error until someone typed the secret password?



````python
word = input("Enter word: ")

print("You typed:", word)

if word != "python":
    print("Oops, sorry.  Wrong word")
else:
    print("Good work, the word was", word)
````

````python
word = ""
while word != "python":
    word = input("Enter word: ")
    print("You typed:", word)
    if word != "python":
        print("Oops, sorry.  Wrong word")
#prints when the correct word is entered
print("Good work, the word was", word)
````
