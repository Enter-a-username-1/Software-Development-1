## Recap

### Loops
* This unit we looked at two different types of loops - `while` and `for`
* `while` loops look like this:
    ```Python
    while boolean condition:
        # do this code
    ```
* `for` loops look like this:
    ```Python
    for i in range(start, stop, step):
        # do this code
    ```
    - OR
    ```Python
    for item in mylist:
        # do this code
    ```

<br>

### Random
* We also introduced the random library which is really useful 
* ##### To get a random floating point value 0-1 do:
    ```Python
    import random

    num = random.random()
    ```


* ##### To get a random integer from 1-10 do:
     ```Python
    import random

    num = random.randint(1, 10) 
    ```
    

<br>

### `for` vs `while`
* When do we use `for` loops and when is it best to use `while` loops?
* `for` loops are great for:
    - Numerical lists
    - Sequences - lists, strings, tuples, arrays
* `while` loops work best when we don't know when we will exit the loop
    - When the user enters some value
    - When something becomes `True`/`False`
    - Good for menus

````python
count = 0
while count < 6:
    print(count, end = " ")
    count += 1
````
