# Plotting with MatPlotLib

* #### To plot data in Python we can use the `matplotlib` library

* `Matplotlib` is very popular scientific Python library
    - One of the trinity: NumPy, SciPy and Matplotlib

* We import `Matplotlib` with a special alias
    ```python
        import matplotlib.pyplot as plt
        
    ``` 

* Basic plotting using `matplotlib` is pretty straightforward:

    1. `import` the functionality to plot (similar to how we have imported other modules)

    2. Construct a data set (either lists, tuples, or numpy arrays)
    3. Create a **figure "object"** - like a *canvas* for the data to be displayed on
    4. `plot` the data on the figure
    5. Label the figure
    6. Tell python to `show()` the figure
 


<br>

### Example: Plot our temperature data with labels
* We can use `plt.title`, `plt.xlabel`, `plt.ylabel`, `plt.legend`, and `plt.grid`

## Numpy array generation tools - `linspace`, `arange`, etc
Sometimes we don't want to generate our lists/arrays by hand using `np.array()`

* Numpy provides a set of tools to create arrays instead of entering the data by hand (we've already looked at random values)
    - `np.random.randint( low, high, num )` random integer array of num values from low to high-1

    - `np.arange(start, stop, step)` is good when you know the step increment (like `range()`).  Goes to stop-1
    - `np.linspace(first, last, num)` is good when you know the number of samples.  Inclusive of endpoints
    - `np.zeros( num )` make an array of all 0s
    - `np.ones( num )` make an array of all 1s
    - `np.full( size, num )` make an array of any size of one number

### *Arange* and *Linspace* can sometimes provide the same results
* `np.arange()` is best when you know the **step** value
* `np.linspace()` is best when you know the **number** of values

````python
# whole numbers 1-100
print( np.arange(1, 101) )                # step doesn't have to be an int
print()
print( np.linspace(1, 100, 100) )         # linspace will alway give a float
````
