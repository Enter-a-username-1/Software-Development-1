## `if`-`elif`-`else` Flow Control (Conditions and Blocks)

* Flow control statements generally start with a condition (A "yes/no" or "True/False" question).
* We use the `if` statement to start our flow control

* Blocks of code:
    - When dealing with conditional execution, we want our code to be grouped in "blocks".
    - In Python, we use **indentation** to indicate a block of code.
    - Recall that when we use a colon (`:`) at end of control statement, in Jupyter the next line automatically indents.

<br>

### Types of Conditionals

1. **One branch** - If a condition is true, execute a block of statement(s), otherwise skip over it.<br><br>

<center><img src="https://i.postimg.cc/nhPk1d4v/llama-1-choice.png" width=250></center>

```python
    if condition == True:
        # Do this stuff
        
    # back to main flow
```
<br><br>

2. <b>Two way alternative branching</b>, one or the other block will be executed.<br><br>

<center><img src="https://i.postimg.cc/6Q4hbdcF/llama-2-choices.png" width=350></center>

```python
    if condition == True:
        # Do one thing
    else:
        # Do the other thing
        
    # back to main flow
```
<br><br>

3. <b>Multi-way alternative branching</b>, one and only one of the action blocks will execute.<br><br>

<center><img src="https://i.postimg.cc/nz8GyCwr/llama-3-choices.png" width=350></center>

```python
    if condition1 == True:
        # Do stuff for a true condition1
    elif condition2 == True:
        # Do stuff for a true condition2
    else:
        # Do stuff when neither condition1 OR condition2 are true
        
    # back to main flow
```
<br><br>

4. Notes
    - There can be <u>ONLY one</u> `if`, it must come first.
      
    - You can have <u>as many</u> `elif` statements/blocks as needed.

    - There can be <u>ONLY one</u> `else`, it must come last, and it has **NO** condition.

## Chained Conditionals - the `elif`
Often our code needs more than a simple binary (this or that) choice.  Sometimes we have options b, c, d etc.  

We want to have an option for `if...else if...else`
<br><br>
In Python to make this happen we use the `elif` conditional (short for "else if") in our code execution.  

The code will check if the first statement is true.  If not, will move to the next and check if it is true and so on.

```python
    if ( condition ):
        # do this
    elif ( condition ):
        # do this instead
    elif ( condition ):
        # do this instead
    else:
        # if all else fails, do this
        # notice there is no condition on else
```
<center><img src="https://i.postimg.cc/YqPLZBdt/if-elif-elif.png" width=300></center>

> There is <u>no limit</u> to the number of `elif` statement you can have.

#### Example

````python
# check value of a
a = 15

if ( a > 100 ):
    print( "a is greater than 100" )
______ ( a < 20 ):
    print( "a is less than 20" )
else:
    print( "a is between 20 and 100" )
````

<br>

#### Example: Check if x or y is greater

````python
# check if x or y is greater
x = 5
y = 17

if x > y:
    print( "x is greater than y" )
_____ y > x:
    print( "y is greater than x" )
______:
    print( "x and y are equal" )
````

<br>

#### Example: How many legs does the Llama have?

````python
# how many legs does the llama have
legs = eval( input( "how many legs does your llama have? " ) )

if legs > 4:
    print( "weird, your llama has", legs, "legs" )
elif legs < 4:
    print( f"poor llama, must have lost {4 - legs} leg(s) in an accident" )
else:
    print( "your llama has a reasonable number of legs" )
````

## Nested Conditionals
You can also put `if` statements inside other `if` statements.  

```python
if (condition):
    # do this
    
    if (nextcondition):
        # do this
        
        if (another condition):
            # do this too
            
    else:
        # do this
        
elif (condition):
    # do this instead
    
else:
    # do this if you have to
```

<br>

> As you can see <b>Nested Conditionals</b> do give us more options.  
> <br>
> However, even though Python uses <u>indentation</u> to try to make branching clear, nested conditionals quickly become <font color="red">*hard to read*</font>.
    
<br>

#### Avoid nesting `ifs` if at all possible for readability

````python
# nested conditional example
x = int(input( "x: " ) )
y = int(input( "y: " ) )

if x == y:
    print('x and y are equal')
else:
    ____ x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
````

<br>

### Sometimes nested conditionals can be simplified with Boolean operators

##### A Nested Conditional

````python
# we can simplify this nested conditional
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
````

##### A little easier to read

````python
# this is a little simpler to read
if 0 < x ______ x < 10:
    print('x is a positive single-digit number.')
````

##### Even easier to read

````python
# Python provides boolean operators to make this even simpler
if 0 ___ x _____ 10:
    print('x is a positive single-digit number.')
````

##### A simpler version of example from above

````python
# a simpler version of our example above
a = int(input("enter number: "))

if (100 < a):
    print("a is greater than 100")
elif (80 < a <= 100):
    print("a is between 80 and 100")
elif (50 < a <= 80):
    print("a is between 50 and 80")
elif (20 < a <= 50):
    print("a is between 20 and 50")
elif (0 <= a):
    print("a is between 0 and 20")
else:
    print("a is negative")
````
