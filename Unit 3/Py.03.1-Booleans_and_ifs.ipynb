## Bool datatype
       
* Most programming languages have built-in data types to deal with items that are yes/no or true/false
   - Referred to as a <b>bool</b> (short for boolean data type - Named after **George Boole**) <br>
   <img src="https://i.postimg.cc/Y08QwK0W/engraving-George-Boole.jpg" width=200>
   <a href="https://www.britannica.com/biography/George-Boole"><br>
    George Boole.  Britannica.com</a>

<br>

* "Boolean" datatypes are either *1* or *0*
    - Or in our case, *True* or *False* (respectively)
        - NOTE: <font color="red">The capital T and F are important here!</font>
<br><br>
    
* For a Bool in Python **everything is True except**:
    - 0
    - None
    - False


````python
# Let's have a look at the Boolean datatye

MyBool = True
YourBool = False

print( MyBool)
print( YourBool)

print( type( MyBool ) )
print( type( YourBool ) )

print( int( MyBool ) )
print( int( YourBool ) )
````

<br>

#### We can use other types in our Booleans

````python
# Using other types with bool

string = "Python"
string2 = "None"
one_num = 1

none_str = None   # this is a special reserved value
zero_num = 0

print( f"{string:>8}: {bool( string )}" )
print( f"{string2:>8}: {bool( string2 )}" )
print( f"{one_num:>8}: {bool( one_num )}" )

print( f"{str(none_str):>8}: {bool( none_str )}" )
print( f"{zero_num:>8}: {bool( zero_num )}" )
````

### Equality of floats <u>can be</u> tricky
* The last digit of a float may not be accurate after multiple operations.

````python
n1 = 1 / 3
n2 = 10 / 3
n2 = n2 / 10  # should be 1/3 now

print( "n1 == n2", n1 == n2 )
print( "n1:", n1 )
print( "n2:", n2 )
print( "Diff between n1 and n2 is:", abs( n1 - n2 ) )
````

## Logical operators (also called Boolean operators)

* Similar to relational operators, there are three common logical operators.
    - `and`
    - `or`
    - `not`
    
In python we can test to see if a combined statements are true or false.  The commands used are called <b>"Logical Operators"</b>

```Python
    x and y            # check if x and y are BOTH True
    x or y             # check if x or y are True
    x not y            # check if x is True and y is not True
    x and z and not y  # check if x and z are true but not y

    llama_wool and llama_legs  # this will check if the llama has wool and has legs
                               # (a number is considered True)

```

These commands can be combined
```Python
    x > y and x < z   # check if x is greater than y and x is less than z
```

<br>

##### Compound booleans

````python
# compound boolean logical operators

print( "True =", True )
print( "True and True =", True and True )
print( "True and False =", True and False )
print( "False or True =", False or True )
print( "not False =", not False )
print( "True and not False =", True and not False )
````

##### Back to our Llama

````python
# Our Llama and boolean/logical operators
llama_wool = True
llama_legs = 4

llama_wings = 0
llama_eyes = 2

llama_fly = False
llama_carnivore = False

llama_ears = 2
llama_beak = False


print( f"The llama has wool and legs? {llama_wool and bool( llama_legs )}" )
print( f"The llama has wings or eyes? {bool( llama_wings ) or bool( llama_eyes )}" )
print( f"The llama can fly and eats meat? {llama_fly and llama_carnivore}" )
print( f"The llama has ears, eyes and no beak? {llama_ears and llama_eyes and not llama_beak}" )
````

##### Logical operators and numbers

````python
# Logical operators with numbers

x = 5
y = 15
z = 33

print( "z > y and y > x", z > y and y > x )
print( "z < y or y > x", z > y and y > x )
print( "z != y", z != y )
````

## Basic Conditional Execution
Conditional execution is really important because it makes the code change depending on what the data is. 

In Python we use the `if` command and on the next line (tabbed in) we write the conditional code
```Python

if (x > y):
    print("x is greater than y")
  
```
The command after the `if` is called the <i>condition</i>

<br><br>
    
**NOTE:** 
* The `if` command always ends with a `:`
<br>
* **INDENT** the code that will be done if the condition is True.  This is <u>REALLY</u> important in Python.

<br>   

#### For example - if the llama has 4 legs, print out congrats
<center><img src="https://i.postimg.cc/8cwSGZhY/llama.png" width=200></center>

````python
llama_legs = 4

if llama_legs == 4:
    print( "Congrats on your llama's legs" )
````

##### Is x positive?

````python
# is x positive?

x = 5

if x > 0:
    print( "x is positive" )
    
print( "this is the end of the program" )
````
