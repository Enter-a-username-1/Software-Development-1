## 1. XY Points

#### Tasks:
* Enter two x,y points
* Determine if they are:
    *  the same point
    *  on a horizontal (y values the same)
    *  on a vertical line (x values the same)
    *  describing a square (the points are two opposite corners of a square)
    *  describing a general rectangle (the points are two opposite corners of a rectangle)
<br><br>
* Use the following example messages in your output:
    - Both points are the same.
    - The points describe a vertical line.
    - The points describe a horizontal line.
    - The points describe a square.
    - The points describe a rectangle.

```text
Enter two points.
x1 =  4.5
y1 =  9
x2 =  3
y2 =  9
The points describe a horizontal line.
```

````python
# get two x,y points
VarX1 = float(input( "Enter x1: " ))
VarX2 = float(input( "Enter x2: " ))
VarY1 = float(input( "Enter y1: " ))
VarY2 = float(input( "Enter y2: " ))

# tests for relationship
if VarX1 == VarX2:
    if VarY1 == VarY2:
        print( "The points are the same." )
    else:
        print( "The points are vertically aligned." )
elif VarY1 == VarY2:
    print( "The points are horizontally aligned." )
elif (VarX1 == abs(VarX2) or VarX2 == abs (VarX1)) and (VarY1 == abs(VarY2) or VarY2 == abs (VarY1)):
    print( "The points create a square around the orgin." )
elif (VarX1 == abs(VarX2) or VarX2 == abs (VarX1)) or (VarY1 == abs(VarY2) or VarY2 == abs (VarY1)):
    print( "The points create a rectangle around the origin." )
else:
    print( "The points are not aligned." )
````

## 2. How comfortable is your house?

#### Tasks:
* Based on temperature and humidity, determine the comfort level of a house.
    - At temperatures above 70F
        - Humidity less than 20% is comfortable
        - 20% to 50% is muggy
        - Over 50 is unbearable
    - When temperature is 70 or less
        - Humidity over 35 is comfortable
        - Anything else is dry
* Prompt the user to enter the temperature, then the humidity (as percentage), determine and display the comfort level.
* Output will be formatted with one decimal place in the temperature, no decimals in the humidity.

#### <font color="blue">Sample run:</font>
```text
Temperature (F):  65.98
Humidity (%):  32.9
At 66.0 degrees and 32% humidity, you will be dry.
```

````python
# get inputs
inputTemp = float( input( "Enter temperature in Fahrenheit: " ) )
inputHum = float( input( "Enter humidity (0-100): " ) )

#first outputs
print("Temperature(F): " + str(inputTemp))
print("Humidity(%): " + str(inputHum))
# temp over 70
if inputTemp > 70:
    # hum over 50
    if inputHum > 50:
        print( "At " + str(inputTemp) + " degrees and " + str(inputHum) + "% humidity, the humidity is unbearable." )
    elif 20 <= inputHum <= 50:
        print( "At " + str(inputTemp) + " degrees and " + str(inputHum) + "% humidity, it very muggy." )
    else:
        print( "At " + str(inputTemp) + " degrees and " + str(inputHum) + "% humidity, it is comfortable." )
# temp 70 or less
else:
    if inputHum > 35:
        print( "At " + str(inputTemp) + " degrees and " + str(inputHum) + "% humidity, it is comfortable." )
    else:
        print( "At " + str(inputTemp) + " degrees and " + str(inputHum) + "% humidity, it is dry." )
````

## 3. Quiz Program
Lets make a program that asks a 4 questions, lets the user know if their answers are correct or not, and tallies their score. 

#### Some sample questions:
* What is the capital of Russia?
* What is 2^3?
* Who wrote 'Romeo and Juliet'?
* What is the largest planet in our solar system?
* What is the chemical symbol for water?

#### Tasks
* Ask each question
* Detemine if the answer is right or wrong
* Tally score
* Repeat for each question
* Print final score
* Print whether they passed or not

<br>

#### <font color="blue">Example run of program</font>
```text
    Welcome to the Quiz! Answer the following questions:

    1. What is the capital of Russia? Moscow
    Correct!

    2. What is 2^3: 9
    Wrong! The correct answer is 8.

    3. Who wrote 'Romeo and Juliet'? Shakespeare
    Correct!

    4. What is the chemical symbol for water? H
    Wrong! The correct answer is H2O.

    Your final score is 2 out of 4.
    Good job! You passed the quiz.

```


````python
# initialize score
PlayerScore = 0

# ask questions
print("Welcome to the quiz! There are 4 questions total. You need to answer at least 3 correctly to pass.")
print("enter your answers below in lowercase.")
input1 = input("What is the capital of South Dakota? ")
if input1.lower() == "pierre":
    PlayerScore += 1
    print("Correct!")
else:
    print("Wrong. The correct answer is Pierre.")
input2 = input("What is 5 + 7? ")
if input2 == "12":
    PlayerScore += 1
    print("Correct!")
else:
    print("Wrong. The correct answer is 12.")
input3 = input("What is the boiling point of water in Fahrenheit? ")
if input3 == "212":
    PlayerScore += 1
    print("Correct!")
else:
    print("Wrong. The correct answer is 212.")
input4 = input("What planet is known as the Red Planet? ")
if input4.lower() == "mars":
    PlayerScore += 1
    print("Correct!")
else:
    print("Wrong. The correct answer is Mars.")

# calculate final score
PercentScore = (PlayerScore / 4) * 100

# print final score and whether they passed
print("Your final score is: " + str(PercentScore) + "%")
print("You answered " +  str(PlayerScore) + " out of 4 questions correctly.")
if PlayerScore >= 3:
    print("Congratulations! You passed.")
else:
    print("Sorry, you did not pass.")
````
