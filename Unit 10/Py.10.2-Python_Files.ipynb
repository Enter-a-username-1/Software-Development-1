## When we use a file in Python, we have to tell it:
1. The **filename**
2. **How** to open it
    ```python
    file_object = open("Filename", "Access mode")
     ```
3. **What** to do with the file
4. **Close** it when done

<br>

### 1. The **filename**
Just a string - "myfile.txt" - with two parts
* A filename
* A 3-4 letter file extension - this is usually hidden in Windows

<br>

### 2. **How** to open the file? aka Python **Access modes** 
- `'r'` is for reading only
- `'w'` is for writing only (overwrites existing file) - `'r+'` for reading and writing
- `'a'` is append only (`'a+'` is for append and read too)



<br>

### 3. **What** to do with it? File Functions

#### There are many functions to **read files** in Python

- `myfile.read()`
- `myfile.readline()`
- `myfile.readlines()`

<br>

#### There are also many functions to **write files** in Python

- `myfile.write()`
- `myfile.writeline()`

<br>

### 4. Finally, when we are done, we will `close` the file or even better, use the `with` statement


<br><hr><br>

## 📖 How to read from a file in regular Python
* Use `open` and `r` access mode

* Use one of the **read functions** with the "file object"

* It is important to `close` the file when done - file doesn't get stuck
* A **better way** is to use the `with` statement
    * It is easy to forget to close a file
    
    * Or if an error occurs, our file is stuck open

<br>

#### Read the whole file as a string using `.read()`
* use `with` and `open(filename, 'r')`
* once open, we need to create a file object with the `as` keyword


````python
# open the file and create file object "myfile"
with open('text.txt', 'r') as myfile:
    
    wholefile = myfile.read()   # read all lines into one string

    # print file contents
    print(wholefile) 
````

<br>

#### Read one line of the file at a time using `.readline()`

````python
# open the file
with open('text.txt', 'r') as myfile:
    oneline = myfile.readline()   # read one line

    # print single line only
    print(oneline)
````

<br>

#### Read all lines into a list using `.readlines()`

````python
# open the file
with open('text.txt', 'r') as myfile:
    list_of_lines = myfile.readlines()  # read all into a list

# print list of lines
print(list_of_lines)
````

<br>

##### Use a `for` loop to go through every line

````python
with open('text.txt', 'r') as myfile:
    for line in myfile:
        print(line)
````

<br>

#### Notice everyline has a newline after it
* This is because it reads a `\n` from the file

* Use `line.strip()` to remove <u>any</u> leading and trailing whitespace
    > This is what your read_text_file function should do in Lab Adventure 6B

````python
with open('text.txt', 'r') as myfile:
    for line in myfile:
        line = line.strip()  # remove trailing \n
        
        print(line)
````

### 1. Read Numbers and Compute a Total

Write a function `sum_numbers(filename)` that:
* Reads a file 
* Each line contains a single integer
* Converts each line to an int (be sure to remove \n)
* Computes and prints the sum of all numbers
* Call with the file called `numbers.txt`

````python
def sum_numbers(filename):
    """Read integers from a file (one per line) and return their sum."""
    

    # initialize total
    total = 0

    # TODO: open the file using a with-statement
    with open(filename, 'r') as myfile:

        # TODO: read each line
        for line in myfile:
            # TODO: convert the line to an int
            number = int(line.strip())
            # TODO: add it to total
            total += number


    return total


# Example call
print("Total:", sum_numbers("numbers.txt"))
````

Write a function `count_lines(filename)` that:

* Counts how many lines it contains
* Returns the count
* Call with file called `text.txt`

**Example behavior**:  
If the file has 12 lines, the function should only return `12`.

````python
def count_lines(filename):
    """Return the number of lines in the given text file."""
    
    
    # TODO: loop over the lines and count them
    
    
    # initialize count
    count = 0

    # TODO: open the file using a with-statement
    with open(filename, 'r') as myfile:
        for line in myfile:
            count += 1

    return count


# Example call
print("lines in text.txt:", count_lines("text.txt"))
````

## 📝 How to Write to a file in regular Python
* Not only can we **read** from files we can **write** to files
* We use `w` or `a` for our "access mode"

<br>

#### Using `w` **"write"** access mode
* Opens file for writing
* Creates the file if it does not exist
* **Overwrites** existing files

````python
with open('newtext.txt', 'w') as myfile2:  # open the file newtext.txt to write to it
    myfile2.write("This is some text")     

print( "writing to newtext.txt complete" )

## take a look at the file we just made
````

````python
with open('newtext.txt', 'w') as myfile2:  # open the file newtext.txt to write to it
    myfile2.write("This is newer text")    # we need \n in our file
    myfile2.write("And here is a second line")

print( "Second writing to newtext.txt complete" )
````

<br>

#### Using `a` **"append"** access mode
* Opens file for writing
* Creates the file if it does not exist
* **Appends** to the end of existing files

````python
with open('newtext.txt', 'a') as myfile2:  # open the file newtext.txt to append to it
    myfile2.write("This is my first line\nThis is my second line\nThis is my last line\n")  # we need \n in our file

print( "appending to newtext.txt complete" )
````

````python
# or write this way
with open('newtext.txt', 'a') as myfile2:  # open the file newtext.txt to append to it
    myfile2.write("This is my first line\n")
    myfile2.write("This is my second line\n")
    myfile2.write("This is my last line\n")

print( "appending to newtext.txt complete" )
````

# 📈 Reading and printing CSV files
* Many data files come to us as **"Comma Separated Files"** or **CSV** files as NumPy arrays
```text
    name,address,phone
    Arthur,16 Camelot Way,42150000

    day,high,low
    1,55,13
    2,60,38
    3,43,22
```
<br>

### We have several options for reading CSV files

1. Read as strings (readlines) and then separate on our own (tedious)
2. Use the `CSV` library to process (can be difficult - we are going to skip)
3. Use `NumPy` (much simpler)
4. Use other libraries we've never heard of (SciPy, pandas, etc) - can be easier or harder

<br><br>

### 🥵 With `open` and `close` (hard way)
1. Read the file
2. Parse the comma separately values
3. Convert to NumPy array

<br>

##### Let's look at the CSV file first!!!!!
* What type of data?
* What is the separator aka "**Delimiter**"?

<br><br><br>

##### Let's read this into a NumPy array

````python
import numpy as np

file_name = "sales.csv"
mylist = []

myarr = np.zeros( (1, 5) )  # initialize a zeros array 1x5

with open(file_name, 'r') as salesfile:
    for line in salesfile:
        line = line.strip()            # remove whitespace
        line = line.split(',')         # split into list separated by commas

        mylist.append(line)

myarr = np.array(mylist)
print(myarr)
````

<br>

### 😅 This is much easier to do with NumPy's `np.loadtxt` function
* Just one line of code

````python
import numpy as np

# load file sales.csv, use ',' as the delimter, datatype is 'str'
salesarr = np.loadtxt('sales.csv', delimiter=',', dtype='str')

print( salesarr )
````
