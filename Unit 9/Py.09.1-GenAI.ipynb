## Turing Test
The most famous test of AI is called the <b><i>"Turing Test"</i></b> proposed by the <br>
British mathematician and computer scientist **Alan Turing** in 1950 in his seminal paper,<br>
"Computing Machinery and Intelligence." It is a much debated test.

<center><img src="https://i.postimg.cc/3Jv4p6st/Turing-Diagram-159676-geeksforgeeks.png" width=600><br>
<font size=2>Source: https://www.geeksforgeeks.org/artificial-intelligence/turing-test-artificial-intelligence/</font>
</center>

<br>

> Alan Turing predicted that by the year 2000, a computer would be able to play the imitation game so well that an average interrogator would have no more than a 70% chance of correctly identifying the machine after five minutes of questioning. However, no computer has yet met this standard.

> In 1990, New York businessman Hugh Loebner announced a $100,000 prize for the first computer program to pass the Turing Test. Despite significant advances in AI, no AI program has yet passed an undiluted Turing Test.

### Types of AI

<br>

* **Machine Learning**
    * Learning from information without explicit programmed instructions
    * Can be guided or unguided

* **Deep Learning**
    * Subset of ML
    * Uses a "Neural Network" mimicking the brain
    * "Deep" because there are multiple layers

* **Generative AI**
    * Large Language Model (LLM)
    * Tokenization (like autocomplete) - <a href="https://platform.openai.com/tokenizer">OpenAI's tokenizer</a>
    * Predictive based on patterns/statistics
    * Can summarize content or generate new content
        * Like an infant - see patterns and mimic in acceptable ways.  Some times generates "new" words or sentences in funny ways
        * Like music - all notes exist. There are acceptable patterns.  Can regurgitate them in new ways

> NOTE: AI does not have emotion (currently).  We give it human terms and associate speech with humanity, but it is mimicking. It "seems" human.
