### Chain of Thought (CoT)

* Break a problem down step-by-step
* See the logic of each step
* Helps the GenAI system process better

##### Prompting Example:
* "In my garage, I pick up a small ball and grab a small box that is missing the bottom.  I walk into my office and I put the small ball into the small box.  Then I take the small box with me to the postal office.  Here I put the small box into a bigger box and send it to my friend in New York.  Where is the ball now?"



<br><br>


### TIC Framework

A very simple framework to remind you of the important components of prompting

* **Task** - goal you want the AI to achieve
* **Instruction** - clear, specific directions on how to accomplish the task
* **Context** - background, persona, target audience etc

##### Example:
* Task: Summarize the key themes and messages in the novel "To Kill a Mockingbird."
* Instruction: Provide an analysis highlighting at least three major themes, such as racial injustice, moral growth, and empathy. Include examples from the text to support the analysis.
* Context: The summary will be used as part of a high school literature review, so it should be accessible and relevant to students familiar with the novel.



<br><br>

### CASE Framework

Similar to TIC, but adds a concrete example - maybe a report, graphic, layout - to guide the LLM's output

* **Context** - background and persona
* **Action** - specific task and goal.  Step-by-step instructions are helpful
* **Specification** - detailed guard rails on tone, format, structure, style, etc
* **Example** - concrete example of how you want the LLM to respond

<br>

##### Bad Example: no CASE
* **Prompt:** give me a blog post about cars

<br>

##### Good Example: with CASE
* **Prompt:** 
    * ***Context***: you are a automotive engineer with strong technical skills, you are famous astrophysicist Neil DeGrasse Tyson etc.
    * ***Action***: create 2 emails to send to potential employers in the automotive industry
    * ***Specification***: the tone is lighthearted and the text must be 3 paragraphs with a strong introduction, make the whole thing less than 100 words
    * ***Example***: attached is an example image, report, website, etc.



<br><br>

### CO-STAR Framework

<center><img src="https://i.postimg.cc/fRftD6nq/prompteng-co-star-tree.jpg" width=600><br>
<font size=2><a href="https://towardsdatascience.com/how-i-won-singapores-gpt-4-prompt-engineering-competition-34c195a93d41/">Source: https://towardsdatascience.com/</a></center><br></font>

* **Context** - Why do I want the model to do this?
* **Objective** - What do I want it to do?
* **Style** - What style should it write?
* **Tone** - How is it communicated?
* **Audience** - Who is the intended audience?
* **Response** - How to structure the output?

<br>

##### Bad Example: no CO-STAR
* **Prompt:** write a text message to a friend on vacation
* **Output:** Hey [Friend's Name]! 😊 Hope you're having an amazing time on vacation! What’s been the highlight so far? Can't wait to hear all about your adventures. Miss you! 🌴✨

<br>

##### Prompting Example: using CO-STAR
* **Prompt:** write a text message to my friend
    * Context: my friend is traveling in France and has been gone for 6 weeks now
    * Objective: a friendly text message
    * Style: relaxed and informal
    * Tone: friendly and missing them
    * Audience: childhood friend who is in college
    * Response: give me 5 sentences
* **Output:** Hey [Friend's Name]!  Wow, I can’t believe it’s been 6 weeks since you left! I hope you’re having the time of your life, eating all the pastries and soaking up the sun. Things here are just the same, but it’s definitely quieter without you around. Can’t wait to catch up and hear all your stories when you get back! Miss you! ❤️✨

### 💡 Prompting Tips
1. Be ridiculously specific — Instead of “young professionals,” say “25–32 year old software engineers in their first job after college”

2. Use concrete examples — Don’t just say “conversational tone” — show a sample sentence
3. Iterate like crazy — Use initial outputs to refine your prompts; it might take 2–3 tries to get perfect results
4. Build a prompt library — Save your successful prompts in a document or tool like Notion for reuse
5. Mix and match — Don’t be afraid to combine elements from different frameworks when needed

### Always be transparent with AI use

1. **Acknowledge AI Assistance<br>**
Always acknowledge the use of AI tools in your work. This transparency is crucial for maintaining academic integrity.

2. **Include AI in Your References<br>**
When citing AI-generated content, include the tool's name, version, and the date/history of interaction. For example: OpenAI. (YEAR). ChatGPT (DATE, YEAR) [Large language model]. Retrieved from [URL].
3. **Combine AI Citations with Traditional Sources<br>**
Use AI tools to complement, not replace, traditional academic sources. Ensure your work includes a balanced mix of AI-generated insights and peer-reviewed literature or primary sources.
4. **Be Mindful of Accuracy and Bias<br>**
Verify the information provided by AI tools. AI models can sometimes generate content that is inaccurate or biased. Cross-reference with reputable sources.
5. **Paraphrase and Quote Appropriately<br>**
If you directly quote AI-generated text, use quotation marks and provide an in-text citation. For paraphrased content, ensure it is substantially reworded and still cite the AI tool.
6. **Avoid Over-Reliance on AI<br>**
Use AI tools to enhance your own understanding and creativity. Avoid over-relying on AI to ensure your work maintains a personal academic voice.
7. **Respect Privacy and Ethics<br>**
Do not input sensitive or confidential information into AI tools. Always adhere to ethical guidelines regarding data privacy and security.
