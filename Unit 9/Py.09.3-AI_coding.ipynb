## 1. Write a program that uses functions to compute mean, median, and mode for a dataset using **native Python <u>only</u>**.  
* Definitions:
    * Mean - average
    * Mode - most frequent value
    * Median - middle value when sorted
* Use AI to write the mode function
* Come up with examples to test that the whole code works
* What are some odd (boundary) cases?
    * Do you need to "refactor" your functions?

#### Starter code for `mean` and `median`

```python
# starter code
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2

    # find the middle number and return it
    if n % 2 == 0:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2
    else:
        return sorted_data[mid]
```
