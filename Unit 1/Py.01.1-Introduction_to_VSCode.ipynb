### Running Cells
* In either case, you can run the cell in one of four different ways:
    1. Click the **"Run"** button to the left of the code cell
    2. Click the **"Run"** button in the floating toolbar above the cell
    3. Click **Shift + Enter** (this will just run the cell and go to the next cell - or create a new cell if there isn't one following)
    4. Click **Alt + Enter** (on Windows) or **Option + Enter** on Mac (both will run the cell and create a new cell below)
* You can also run all the code in the notebook by pressing the **"Run All"** button in the toolbar.
