### Data Types

#### How do we know "Hello Monty" is a string?  And what does "it's a string" even mean?
  - In programming "data types" are <font color="green">important</font>.  
      - Data types change how data is stored, used or what actions can be performed <br><br>
  - We can use the built in `type( )` function to see what data type we're dealing with like this:
    ```python
      type(variable_name)
    ```

<strong>Print Variable Type</strong>

````python
# We can print the type of a variable directly

thisvar = 9.997

print( float( thisvar ) )
````

#### What about other types?

````python
# Let's create a few variables and display their types

x = 1
y = 1.
z = 4 + 6j

print( type(x) )
print( type(y) )
print( type(z) )
````

#### Looks like:

## Rules for defining variables

### Main Rules
1. A variable can **ONLY** contain alpha-numeric values and underscores 
    * e.g., F3Test or F3_Test or f3_test, etc.
2. A variable must start with a letter or an underscore
    * Thus a variable **CANNOT** start with a number
  
### Also Important 
1. A variable is case sensitive! 
    * e.g., F3Test is NOT the same as f3Test and F3test and f3tEst, etc.
1. Do not use Python keywords or functions as names 
    * `print`, `input`, `math` etc

## Strings and string manipulations
We've seen strings above, let's get a bit more familiar with them

#### Strings are anything between double quotes (`"`) or single quotes (`'`)
* They can be long or short, one letter or hundreds
* **IMPORTANT**: The <b>+</b> and <b>*</b> operators act differently on strings than on numbers

````python
# some new variables

str1 = "Spam"
str2 = "eggs"
str3 = "bacon"
str4 = "and spam"
````

<strong>Plus `+` Operator</strong>

````python
# the + operator concatenates the strings 

CompleteStr = str1 + " " + str2 + " " + str3 + " " + str4

print( CompleteStr )
````

<strong>Duplicate `*` Operator</strong>

````python
# the * operator replicates the string

ThreeSpams = str1 * 3

print( ThreeSpams )
````

````python
# the * operator replicates the whole string

FourEggs = ( str2 + "." ) * 4

print( FourEggs )
````

<br>

### What about mixing types?

````python
# What about mixing number and string types?

num_eggs = 56
MyString = num_eggs + "Eggs"

print( MyString )    # this will give an error
````

<br>

### Oops!  What just happened????
<br>

>#### <font color="red">We just got our first error!!!</font> aka the <font color="blue">"big kaboom message"</font>

<br>

**The Problem:** You can't mix numerical types with strings
- We need to "re-cast" the integer to a string
- Use the `str()` function
    

````python
# The str() function will type cast (change) the numerical value to a string 

MyString = str( num_eggs ) + " Eggs" 

print( MyString )
````

````python
# What about other casting? 

MyOtherString = "Eggs" * 2.5

print( MyOtherString )  # this will also give an error
````

> **Error:** Looks like we can't multiply a `string` by a `float`

````python
# Let's cast the float to an int and retry 

MyOtherString = "Eggs" * int( 2.5 )

print( MyOtherString )
````

> When you cast as an `int` it drops the decimal

<hr>

## Type Casting
* Type casting means to change the "type" of a variable
* We can typecast between different types when appropriate
* We can use:
    - `int()`
    - `float()`
    - `str()`
    - or other datatypes we will learn later
* This will sometimes change the data - ie `int(2.5)` becomes `2` and loses precision 

````python
# different ways to cast variables 

# cast in the print statement

x = 52.769

print( int( x ) )
````

````python
# cast first then print

x = int( x )

print( x )
````

````python
# convert int y to different data types

y = 2
print( y, "is a", type(y) )

y = float(y)
print( y, "is now a", type(y) )

y = str(y)
print( y, "is finally a", type(y) )
````

````python
# we can display scientific notation

z = '1.23e5'
print( z, "is a", type(z) )

z = float(z)
print( z, "is now a", type(z) )

z = int(z)
print( z, "is lastly a", type(z) )
````

## Final Important Topic: Getting User input
* It is very important for our programs to get data
* We can "collect" data from the user with the `input( )` function
    - By default `input( )` gives us the data as a  **string**
    - We say that "`input()` **returns** a string"
  > we may need to cast this to use it

````python
# Get a users age and output 2 times their age 

age = input( "Please enter your age: " )

print( age * 2 )

print( type( age ) )
````

> **What happened here????**
> - We need to typecast `age`

````python
# change age to an int before multiplying

age = int( age )

print( age * 75 )

print( type( age ) )
````

> **We can typecast on the same line when we get the input**<br><br>
> lets try a float this time

````python
# Get a users age and output 2 * age 
# all in one step

age = float( input( "Please enter your age: " ) ) # cast the input to a float 

print( age * 2 )
````

> There is nothing wrong with doing it in two steps.
> - Remember "simpler is better"

````python
# Get a users age and output 2 * age 
# in two steps

age = input( "Please enter your age: " ) 

age = int( age ) # now cast the age to a float 

print( age * 2 )
````

### `eval()`
* We can also use the special `eval( )` function to let python determine the "most appropriate" type
* `eval()` is smart enough to determine whether the value should be an int or float
* **NOTE:** `eval()` does present a security risk - we will address that later

````python
# Get a users age and output 2 * age 
age = eval( input( "Please enter your age: " ) ) # cast the input to either an integer or float

print( type( age ) )

print( age * 2 )
````

#### We can also make our output a little nicer using `,`

````python
print( "Your age is", age, "years" )

print( "2 times", age, "years is", age * 2 )
````

> The `,` operator, separates our print statements for us (and adds a space too)
