````python
# insert AI Generated function here
import os

def read_text_file(filename):
    try:
        with open(filename, 'r') as f:
            lines = tuple(line.strip() for line in f if line.strip())
        return lines
    except Exception as e:
        print(f"   An error occurred {e}")
        return tuple()
````

````python
# insert AI Generated function here
import numpy as np

def count_adjacent(grid, row, col):
    count = 0
    rows, cols = np.shape(grid)
    for dr in [-1, 0, 1]:
        for dc in [-1, 0, 1]:
            if dr == 0 and dc == 0:
                continue
            r, c = row + dr, col + dc
            if 0 <= r < rows and 0 <= c < cols:
                if grid[r][c] != '.':
                    count += 1
    return count
````

````python
# test function DO NOT MODIFY
def test_count_adjacent(grid, tests):
    passed = True
    size = np.shape(grid)[0]
    for row in range(size):
        for col in range(size):
            result = count_adjacent(grid,row,col)
            if result != tests[row*size+col]:
                print(f"Test failed at ({row},{col}) {tests[row*size+col]} != {result}")
                passed = False
    if passed:
        print("All tests passed")    
````
