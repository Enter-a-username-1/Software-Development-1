## Scoping Inside Functions
* Variables in functions act slightly differently
* They are not usually accessible after the function completes 

* Given the following function

```python
        def addVals(p1, p2):
            result = p1 + p2
            return result

        n1 = 10
        n2 = 20

        total = addVals(n1, n2)
        print(total)
```

* If we try top `print(p1)` outside the function we will get an <font color='red'>**ERROR**</font>

<br>

> ### Things to note:
>
>* The names of our arguments `n1 & n2` are <b>different</b> than the parameters inside the function - `p1 & p2`
>
>    - When we "**pass**" the arguments into the function, the values get re-assigned to the variable names p1, p2 (this is how the function uses them)<br><center><img src="https://i.postimg.cc/0rRLjhBg/passing-values-to-fn.jpg" width=400><figcaption><font size=2>passing variables</font></figcaption></center>
>
>    - This is sometimes referred to as **"Pass By Value"**<sup>1</sup>
>
>    - **<font color="red">MUST take caution here!!!</font>** (it's best NOT to re-define the function parameters inside the function)<br><br>
>
<font size=1><sup>1 </sup>Technically Python uses **"Pass By Object Reference"** to pass arguments to functions.  Read more about this <a href="https://www.geeksforgeeks.org/is-python-call-by-reference-or-call-by-value/">HERE</a>

<br>

### What happens if we change a parameter variable inside the function?

````python
# bad version of addVals
def badAddVals(p1, p2):
    p2 = 50               # Why did we do this?  What were we thinking?
    result = p1 + p2
    return result
````

<br>

#### Let's call the function now and see what happened to the arguments we pass in

````python
n1 = 10
n2 = 20

total = badAddVals(n1, n2) # we expect 10 + 20 = 30

print(total)
````

> ##### We expect to see 10 + 20 = 30, but that's not the result
>
> **Oh No!!!!** Modifying a parameter inside the function changed our expected result


#### What if we try to `print(p1)` after the function call?

````python
total = badAddVals(n1, n2) # we expect 10 + 20 = 30
print(total)
print(p1)
````

### Function Scope
* In Python: anything defined inside the function is *local only* to the function.
* Only the function has access to it.
    - In fact, the variable **only exists** during function execution
        - Once the function is finished executing, **that variable no longer exists**
        
* This is called "Function Scope"
