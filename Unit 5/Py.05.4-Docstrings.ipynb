# Tuple Assignments

## What can a function return?

* #### Pretty much anything!
    - Scalar values
    - Floats
    - Complex numbers
    - Strings
    - Booleans
    - Tuples and Lists
    - Numpy arrays

* BUT technically only **one** thing


### What is a Tuple Assignment?
* Strictly speaking, functions can **only** return <font size=8>one</font> thing
    
* But when returning multiple values, they get packed into a structure called a **tuple** (it is like a list that cannot be modified)
    * This is called **"Tuple Assignment"**

    * We **comma** separate the return values

    ```Python
        def get_values():
            # do some code here
            return j, k, l
    ```

    * Now, we can unpack the **tuple return** with multiple destination variables
    ```Python    
        a, b, c = get_values()
    ```

    * So variable
        * $j$ gets copied to $a$
        * $k$ gets copied to $b$
        * $l$ gets copied to $c$
    

<br>

### Example: Get coordinates for a 3D point

````python
# get a 3D point's coordinates
def get_xyz( ):
    x = eval(input("X: " ))
    y = eval(input("Y: " ))
    z = eval(input("Z: " ))
    return x, y, z            # return multiple values as a datatype called a tuple



# test the function
x1, y1, z1 = get_xyz()
print(f"({x1}, {y1}, {z1})")
````

### Let's make a function to print these coodinates
* Don't print a newline at the end

````python
# print x,y,z coodinates inside parenthesis, don't do newline
def print_point( x, y, z ):
    print( f"({x:.0f}, {y:.0f}, {z:.0f})", end='')
    
````

<br>

#### Call the function and print results

````python
# get the points from the user
x1, y1, z1 = get_xyz()  
print()
x2, y2, z2 = get_xyz()


# now print them nicely
print("The two points are ", end='')
print_point(x1, y1, z1)
print(" and ", end='')
print_point(x2, y2, z2)
````

So you see we can return more than one value if we **comma separate** them
* In Python this makes **one tuple** - *more in the next unit of tuples*

## Docstrings

#### It is very important to document our function code
1. It explains what the function does
2. It helps others (and future you) use the function correctly
3. Supports testing and debugging
4. Built-in tools (like VsCode) use them
5. Looks more professional


#### Most built-in functions are documented
* Use `help` to access the documentation for `print`

````python
help(print)
````

### In Python functions, we document with a thing called a **Docstring**
* This is the first line in a function
* Usually separated by triple quotes `"""`
* This is a one line or multi-line string describing:
    - What the function does
    - Input parameters
    - Return values
    - Example usage
    
* For example
    ```python
    def get_max( num1, num2 ):
        """this function will return the max of two values
        Parameters: num1 and num2
        Return: the max value
        Example: mymax = get_max( n1, n2 )
        """
        if num1 > num2:
            return num1
        else:
            return num2
    ```
* This allows us to use `help(getMax)` to find out about this function

##### Example: docstring with `help` function

````python
def get_max( num1, num2 ):
    help(get_max)

    if num1 > num2:
        return num1
    else:
        return num2
````

````python
print(get_max)
````

##### Example: hypoteneuse function using Pythagorean Theorem

````python
def hypotenuse(a, b):
    """
    Returns the length of the hypotenuse
    of a right triangle with legs a and b.
    ________:
        a: one leg of triangle
        b: other leg of triange
    _______: hypoteneuse
    _______: result = hypoteneuse(3, 4)
    """
    return (a**2 + b**2)**0.5

help(hypotenuse)
````

#### Example: grade letter and pass fail functions
* every function has a **docstring**
* also notice that the commenting in the functions themselves

````python
##################################################
## letter_grade function
##################################################
def letter_grade(grade):
    """This is a single function doctring for this 
    function that gives a letter grade

    Parameters: grade - users score in percent
    Return: Letter grade string
    Example: letter = letter_grade( grade )
    """
    
    # check percentage
    if grade >= 94:
        return "A"
    elif 90 <= grade <= 93:
        return "A-"
    elif 87 <= grade <= 89:
        return "B+"
    elif 84 <= grade <= 86:
        return "B"
    elif 80 <= grade <= 83:
        return "B-"
    elif 77 <= grade <= 79:
        return "C+"
    elif 74 <= grade <= 76:
        return "C"
    elif 70 <= grade <= 73:
        return "C-"
    elif 67 <= grade <= 69:
        return "D+"
    elif 64 <= grade <= 66:
        return "D"
    elif 60 <= grade <= 63:
        return "D-"
    else:
        return "F"   # all other grades will be F
    
    
````

````python
##################################################
## pass_fail function
##################################################
def pass_fail(grade):
    """This function will determine the 
    letter grade and pass/fail of a grade

    Parameters: grade value as percent
    Return: grade letter and pass/fail determination as *tuple*
    Example: let, passed = pass_fail(mygrade)
    """
    
    # get the letter grade
    letter = letter_grade( grade )
    
    # determine if pass or fail
    if letter == "F" or letter == "D-":
        return letter, False                # return as tuple
    else:
        return letter, True                # return as tuple
    
````

````python
############# MAIN CODE #############################   

# get user's score
mygrade = eval(input("What is your grade: "))

# get letter grade and pass/fail
let, passed = pass_fail(mygrade)

# print score and results
print(f"With a score of {mygrade:.0f} you earned the letter grade {let} and you {passed}" )
````

### 1. Given the `get_valid` function below, write a docstring
Be sure to include
* Description
* Parameters
* Returns
* Example Usage
* > Note: this function's job is to get a valid user input between two values


````python
def get_valid(x, y):
    """
    {description}

    Parameters: take in two values, x and y, that represent the lower and upper 
    bounds of a valid number
        
    Returns: a valid number between x and y
    Example: x = 10
             y = 20
             num = get_valid(x, y)
        

    """
    
    prompt = f"Enter a value between {x} and {y}:"
    num = eval(input(prompt))
    
    while num < x or num > y:
        print(num, "is invalid")
        num = eval(input(prompt))
        
    return num
````
