### Let's Try this function

##### First we **define** the function

````python
# define the function
def greeting_function(username):
    print(f"Hello {username}!  Welcome to CSC115!")
````

##### Notice nothing happens until we **call** the function

````python
# function call
greeting_function("Bob")
````

We can call with a variable

````python
# call with a variable
usr = "KArthur88"
greeting_function(usr)
````

We can call the function multiple times

````python
# call the function multiple times
usr1 = "Galahad"
usr2 = "Robin"
usr3 = "Lancelot"

greeting_function(usr1)
greeting_function(usr2)
greeting_function(usr3)
````

#### Example: Write a function called `cyl_vol` to calculate the volume of a cylinder with a given radius and height. 
* The function will compute and print the volume.
* Recall, given the radius ($r$) and height ($h$) of a cylinder, we can compute the volume via:
    >$V = \pi r^2 h$

````python
import math

def cyl_vol( rad, h ):
    vol = math.pi * rad**2 * h
    print("Volume is", vol)
    
````

* Now call the function a few times, using different arguments. 

````python
cyl_vol(4,10)
cyl_vol(2,5)
cyl_vol(3,7)

# try many others
````

### **Function Tips:**
- Function definitions nearly <u>always</u> go at the **top** of the program, right after `imports` and **constants**.

- `def` defines the functions
    - `def` does NOT actually run (call) the code - it simply let's Python know a new function exists and what it is supposed to do.

- **Calling** the function makes Python run the code inside the definition

- The argument gets assigned to the function's ***local* variable** inside the function, this is referred to as a **parameter**

- Need to **test** the function to make sure it really works the way you are expecting
