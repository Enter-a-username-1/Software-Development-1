#### Example: Max of 3
* Print the largest of 3 values

````python
# function definition
def max_of_three(a, b, c):
    # assume a max
    mymax = a

    if b >= a and b >= c:
        mymax = b
    elif c >= a and c >= b:
        mymax = c

    print(f"The max of {a}, {b} and {c} is {mymax}")


# call function
max_of_three(1, 2, 3)
max_of_three(10, -2, 4)
max_of_three(-8, 14, 7)
````

#### Basic Terminology for Functions:
* In the function defition:
    - **define (def)**: defines what the function will do
    - **parameter**: variables that have arguments assigned to them in your function  
    - **function header**: the def, function name and parameters
    - **function body**: everything after the header - what the function does

* In the function call:
    - **call**: using the function in your program
    - **pass**: how you provide information from your program to a function
    - **argument**: the information you are passing to the function from your program

## `return` statement

As we've already seen, we can have two types of functions: those that `return` a value, and those that don't

#### `return` is very powerful in functions
* It ends the function immediately
    * Similar to `break`
* **"Returns"** a value to the **main** program
* We have to **do something** with the returned value


````python
# function definition
def test_return():
    print("2. Before return")
    return 42     # ends the function immediately

    print("This will never run")
    print("neither will this")


# function call
print("1. Call the function")
return_val = test_return()       # note not every function has arguments
print("3. After return")

print(f"4. Returned value = {return_val}")
print("5. Done with code")
````

## Two types of functions: **Void** (or non-Valued) and **Valued**

<center>

| **Void** Function    | **Valued** Function  |
| ---------------- | ---------------- |
| Does some action | Does some action |
| No value returned | Will `return` a value |
| i.e. `print()`  |  i.e. `input()`<br>`math.cos()` |

</center>

<br>

> ### Simply, if it has a `return` it is a **valued** function

````python
# Void function (prints but returns nothing)
def add_print(a, b):
    return(a + b)

add_print(3, 4)
````

````python
# Valued function (returns a result)
def add(a, b):
    return a + b

x = add(3, 4)
print("Result times 10 =", x * 10)
````

>#### **IMPORTANT:** Do <u>something</u> with the return value

#### Example: Calculate the Cube Root
* Write a **valued** function that returns the cube root of a value
* Name it whatever you would like
* How many parameters would this function have?

````python
# write your function definition here
def cbrt( num ):
    ans = num ** (1 / 3)
    return ans
````

````python
# get input and call your cube root function
num = eval(input("Enter number to cube root: "))

print(f"The cube root of {num} is {cbrt(num)}")
````

### 4. Write a function called `get_valid` to get a valid value between two given numbers. 
* Assume the first number is the lower of the two
* Prompt user to enter a number between two numbers
* Keep prompting until the number is valid
* `return` the valid value from the function
* Print the valid value **in the main code**
* *Is this a void or valued function?*

````python
# get_valid function
# returns a valid number between x and y
def get_valid(x, y):
    while True:
        num = eval(input(f"Enter a number between {x} and {y}: "))
        if x <= num <= y:
            return num
        print("Invalid input. Try again.")

````

````python
# main code
inputBottom = eval(input("Enter First Number: "))
inputTop = eval(input("Enter Second Number: "))
get_valid(inputBottom, inputTop)
````

### 5. Write a function called `factorial` that will return the factorial of the parameter. 
* Recall that $n! = 1 \cdot 2 \cdot 3 \cdot ... \cdot (n−2) \cdot (n−1) \cdot n$
* In the main code, prompt the user for the number to take the factorial of
* Call the function (be sure to do something with the `return` value)
* Print the result like:
```text
    5! is 120
    
    9! is 362880
```
* *Is this a void or valued function?*

````python
# factorial function definition here
# returns the factorial of n
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
````

````python
# main code
# get input
num = int(input("Enter number to find factorial: "))

# call function
result = factorial(num)

# display results
print(f"{num}! is {result}")
````
