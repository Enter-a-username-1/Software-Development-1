## 2. List/Tuple Slicing

Slicing works on lists and tuples **just like in strings**
```Python

```

````python
list3 = [1, 3, 5, 6, 8, 55, 7, 8, 9]
print( list3 )
print( "list3[1:3] ", list3[1:3] )
print( "list3[5:]  ", list3[5:] )
print( "list3[::-1]", list3[::-1] )
print( "list3[::2] ", list3[::2] )
````

### 3.1 Building a List using `append`

It is easy to make a list of a few items.  But what about hundreds?
* We should use a loop
* We need to initialize our list **above the loop**

##### There is a special method `append()` that adds to the end of a list

#### Example: Build a list of the squares of all numbers 0-25

````python
# create a list of the squares of all numbers 0-25

# initialize empty list
my_squares = []    

# loop 
for i in range( 26 ):
    my_squares.___________( i ** 2 )   # add i^2 to the end of the list
    
print( my_squares )
````

<br>

#### To find the location of an item in a list use `index()` method

````python
my_list2 = [1, 4.5, 6, 77, -100, 44, 14, 55.6, 14]

loc = my_list2.________(4.5)  # where is 4.5?
print("4.5 at", loc)

loc = my_list2.index(14)   # where is the first 14?
print("14 at", loc)

loc = my_list2.index(200)  # where is 200 in the list?  Will give an error!
print("200 at", loc)
````

A better way that avoids the error

````python
if 200 _______ my_list2:
    loc = my_list2.index(200)  
    print("200 at", loc)
else:
    print("200 not found")
````

#### Example: remove 'rat' from the list

````python
# remove 'rat' from the spam list
mylist = [ "cat", "bat", "rat", "elephant" ]
print("before remove:", mylist)

if "rat" in mylist:
    mylist._________("rat")
    
print("after remove  ", mylist)
````

> Note how the presence of "rat" was first tested with the `in` operator. <br>
> <br>

````python
somelist = [ 1, 4, 7, 23, 1, 4, 0, 19, 4, 3, 26, 15, 22, 2, 4, -5, 68, 4, 2, 1 ]

print("original list: ", somelist )

# remove every 4 found
________ 4 in somelist:
    somelist.remove(4)   # remove all 4s
    
print()
print("All 4s removed:", somelist )
````

## 4. Sorting a list

#### There is a built-in method called `.sort()`
* Uses a special sorting algorithm called <a href="https://en.wikipedia.org/wiki/Powersort">**powersort**</a>
* Versions prior to Python 3.11 used <a href="https://en.wikipedia.org/wiki/Timsort">**timsort**</a><br>

    > **IMPORTANT:** `.sort()` will change the order of the original list

````python
somelist = [1, 4, 7, 23, 1, 4, 0, 19, 4, 3, 26, 15, 22, 2, 4, -5, 68, 4, 2, 1]
print("original:", somelist)

somelist._______()  # sort and change the original

print("sorted:  ", somelist)

somelist.sort( reverse=________ )  # sort in reverse order

print("reversed:", somelist)
````

<br>

> Notice: to sort in ascending order use `.sort(reverse=True)`


<br>

#### We cannot use `.sort()` on a tuple because tuples cannot be changed (***immutable***)

````python
mytuple = ( -5, 0, 23, 7, 26 )

print(mytuple)
mytuple.sort()

print(mytuple)
````

<br>

> #### When we use `.sort()` it will modify the original order of our list

<br>

#### The function `sorted()` creates a **new** sorted list from any sequence
* It **does not** modify the original tuple/list/string

````python
mytuple = (0, -5, 26, 23, 7)

print("original tuple:", mytuple)
sorted_tuple = ___________( mytuple )

print("sorted list   :", sorted_tuple)
print("original unmodified:", mytuple)
````

## 6. Sequences in Functions
* ***Mutable*** variables like lists act differently when passed to a function
* When we modify the list in the function, it modifies the **original** list

<br>

<center><img src="https://i.postimg.cc/59vqst8Y/06-no-vegas.jpg" width=400></center><br>

#### With lists, what happens **in** the function, IS remembered **outside** the function

* This is because lists are not copied, but sent "by reference" - it sends the location of the original list
* In Python this only happens with mutable datatypes - lists, dictionaries, and sets

<br>

### List as a return value

````python
# get_weights function
# has a list as return value
def get_weights( ):
    
    # make an empty list to add to 
    weights = _____  

    # get first weight
    value = eval(input("Enter weight: "))

    # ask for a weight value and append it to a list
    # this will keep asking until value is blank
    while value > 0:
        weights.append(value)
        
        value = eval(input("Enter weight: "))
                 
    # return list
    return ________


# call the function
all_weights = get_weights( ) 

# print list
print( all_weights )
````

<br>

#### List parameter that gets modified
* Subtract clothing weight from weight
* To modify the list we have to loop over indices<br><br>
* **Notice** we do not return the list
    * Is this *void* or *valued*?

````python
# sub_clothes subtracts clothing weight from list
# list parameter that gets modified
def sub_clothes( weights_list, clothes_weight ):
    
    # must use range and indexing method to loop, 
    # do not use: for i in weights
    # because we need to modify list elements by index
    for i in range( _______(weights_list)): 
        
        # modify the value of weights
        weights_list[______] -= clothes_weight  


## main code and function call
print( "original weight:", all_weights )


sub_clothes( all_weights, 5 )  # sub 5 lbs
print( "minus clothes:  ", all_weights )
````

## 7. Lists & Strings

Lists and strings are not the same thing but are similar
> you can convert a `string` to a `list`

````python
# define string
my_string = 'spam'
print( "string:", my_string )

# the list function breaks strings into individual characters
my_list = ______( my_string )
print( "list:  ", my_list )
````

<br>

### We can `split` a string to make it a list
* We split on a character called a **delimeter**


#### Split a string by spaces

````python
# Default split is on spaces

# Define string
sentence = "Learning new things is fun"
print("original sentence:", sentence)

# Convert the string to a list using split
words = sentence._______()    # split the sentence by spaces
print("split on spaces:  ", words )
````

<br>

#### Split a string on hyphens

````python
# Split on hyphens instead
delimiter = '-'

spam = 'spam-spam-spam'
print("original sentence:", spam)

split_spam = spam._______('-')   # split hyphens instead
print("hyphen split:     ", split_spam)
````

<br>

### `Join` is the opposite of split

````python
wordlist = ['this', 'is', 'another', 'sentence', 'to', 'see']
print("original list:   ", wordlist)

delimiter = ' '

new_sentence = delimiter.______(wordlist)  # join a list together using spaces to separate
print("joined sentence: ", new_sentence)
````
