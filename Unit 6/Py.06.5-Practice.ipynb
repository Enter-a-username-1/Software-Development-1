## A Brief Demonstation of Lists vs Tuples

We have said that tuples are more efficient than lists, but what does this really mean?

Computing efficiency is largely a factor of **resources** used and **time** used

#### So we are going to look at 2 factors:
1. Memory
2. Speed

### Memory efficiency

````python
import sys

# build list and tuple
a_list = [ "Spam", "Eggs", "Sausage", "And", "Spam" ]
a_tuple = ( "Spam", "Eggs", "Sausage", "And", "Spam" )

# find memory size
print( f"list is {sys.getsizeof(a_list)} bytes" )
print( f"tuple is {sys.getsizeof(a_tuple)} bytes" )
print( f"That is a difference of {sys.getsizeof(a_list) - sys.getsizeof(a_tuple)} bytes" )
````

### Speed
* Create a list and tuple of 100,000,000 elements
* There are only slight differences here
* This code may take >1min to run

````python
import timeit

def build_list(n):
    mylist = list(range(n))
    for i in range(len(mylist)):
        a = mylist[i]

def build_tuple(n):
    mytup = tuple(range(n))
    for i in range(len(mytup)):
        a = mytup[i]     

x = 100_000_000
n = 10
time1 = timeit.timeit(lambda: build_list(x), number=n)
time2 = timeit.timeit(lambda: build_tuple(x), number=n)

time1 /= n
time2 /= n

print(f"Average execution time of list over {n} runs: {time1} seconds")
print(f"Average execution time of tuple over {n} runs: {time2} seconds")
print(f"Tuple operation is {time2 - time1} seconds faster")
````

### 1. Given the following list
* Write a function that receives the list as a parameter
* This function will return following as a tuple:
    * `max`
    * `min`
    * `len`
    * `sum`
    * average 
* In the main program print these values
* Also find the index where the max and min occur 
    * HINT: We can use `.index()` on lists which is similar to `.find()` in strings

* Your output will look like:
```text
        max: 220 occurs at index 5
        min: -45 occurs at index 8
        len: 10
        sum: 418
        avg: 41.8
```

````python
def list_stats(lst):
    if not lst:                      # handle empty list just in case
        return None, None, 0, 0, 0
    mx = max(lst)
    mn = min(lst)
    length = len(lst)
    total = sum(lst)
    avg = total / length
    return mx, mn, length, total, avg

#### main code - call function and print below
my_list = [1, 3, 42, 4, 98, 220, 45, 17, -45, 33]

mx, mn, length, total, avg = list_stats(my_list)

print(f"max: {mx} occurs at index {my_list.index(mx)}")
print(f"min: {mn} occurs at index {my_list.index(mn)}")
print(f"len: {length}")
print(f"sum: {total}")
print(f"avg: {avg}")
````

### 4. Using the `append` method, build a list of 20 random numbers -100 to 100 inclusive
* print the max and min of this list
* Be sure to initialize your list above the loop
