<br>

A list is a sequence of values - characters, numbers, strings or a mix of all of these.  
* These are called <b>elements</b> or <b>values</b>

### Lists are made with the **square bracket** operators `[ ]`

```Python
    # lists are made with square brackets
    [ 10, 20, 30, 40, 50 ]
    [ 'holy', 'grail', 'dennis', 'patsy' ]
    [ 'spam', 6.7, 12, -4, "crunchy" ]
```

<br>

````python
list1 = [ 10, 20, 30, 40, 50 ]
print( list1 )
````

````python
list2 = [ 'holy', 'grail', 'dennis', 'patsy' ]
print( list2 )
````

<br>

#### We can also create an **empty** list
* We can add to it later

````python
# an empty list
empty_list = []

print(empty_list)
````

<br>

#### We can use the `list()` function to typecast from one datatype to a list
* similar to how we cast from `float()` to `int()`, etc.
* Recall our `range()` function we used in the for loop?

````python
print( range( 2, 50, 2) )

print( list( range( 2, 50, 2 ) ) )
````

<br>

## Nested Lists
Lists can be <b>nested</b> inside one another

````python
nested = [ 30, 40, list([1, 2, 3]), 50 ]
print( nested )
````

````python
knights = [ 'lancelot', 'galahad', 'boors', [ 'patsy', 'robin' ], 'bedevere' ]
print( knights )
````

<br> 

#### We can mix & match datatypes

* Multiple data types
* Same data type
* Even other lists

````python
# List of strings 
spam = [ "cat", "bat", "rat", "elephant" ]
print( "strings:", spam )

int_nums = [ 1, 2, 3, 4, 5 ]
print( "ints:   ", int_nums )

float_nums = [ 1.2, 2.2, 3.2 ]
print( "floats: ", float_nums )

mixed = [1.2, 55, 'dog']
print( "mixed:  ", mixed )
````

##### We can even make a list of lists

````python
# List of lists
list_of_lists = [ spam, int_nums, float_nums, mixed ]

print( list_of_lists )
````

## Accessing List Items
> List items are accessed using the **index** <u>just like</u> in `strings`

<br>

Use the square brackets operator
```Python
    mylist = [ 0, 1, 2, 3, 4 ]
    print( mylist[1] )
```

#### We access individual elements of the list using **indexes**
- **Recall:** the first element starts at **0** NOT 1
    
<figure>
<center>
<img src='https://i.postimg.cc/mgXRV6qm/list-elements.png' width="600px" />
<figcaption>List access using indexing.</figcaption></center>
</figure>

````python
mylist = [ 40, 61, 22, 43, 84 ]
print("The 2nd element is", mylist[1] )  # 2nd element
````

````python
print("The 5th element is", mylist[4] )  # 5th element
````

> Be careful you don't go **past the end** of the list

````python
# past the end of the 5 element list
print( mylist[5] )    # error
````

<br>

#### Like a string, we can find the length of a list with the `len()` function

````python
length = len(mylist)
print("The length of mylist is", length )
````

````python
# print the last item using len
print( mylist[len(mylist) - 1] )
````

<br>

> ##### Using a **negative index** counts backwards from the end of the list

````python
print( mylist[-1] )
````

<br>

#### To access nested list elements use **two** indices
```Python
    list_name[index_in_outside_list][index_in_inside_list]
```

````python
# print list_of_lists from above
print( list_of_lists )
````

````python
print( list_of_lists[0][1] )        # bat
print( list_of_lists[1][3] )        # 4
print( list_of_lists[2][0] )        # 1.2
print( list_of_lists[3][2] )        # dog
````

## Unlike strings, lists are totally **mutable**
* You can change individual elements of a list
* This is more efficient than creating a whole new list
* But it does make the `list` object **slower** to use than the `string` or `tuple`

````python
mynewlist = [ 'a', 1, 217, 299, "monty", "python", 1999 ]
print( mynewlist )
````

````python
mynewlist[3] = 46578    # change 4th element
print( mynewlist )
````

````python
mynewlist[4] = [ 'King', 'Cape', 'Forest', 'Caspian' ]    # change 5th element
print( mynewlist )
````

<br>

### Recall the `in` operator
* One of the most powerful operators in Python
* you can see if something is in a list using the `in` operator

````python
print( 1 in mynewlist )  # check if 1 is in mynewlist
````

````python
print( 'monty' in mynewlist )  # check if 'monty' is in mynewlist
````

````python
if 299 in mynewlist:
    print("299 found")
else:
    print("299 not found")
````


<br>

#### Be careful using `in` with nested lists

````python
print( mynewlist)
print( 'King' in mynewlist )
````

````python
print( [ 'King', 'Cape', 'Forest', 'Caspian' ] in mynewlist )
````

````python
print( 'King' in mynewlist[4] )
````
