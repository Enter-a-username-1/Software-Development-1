## String Indexing

````python
mystring = "Monty Python"
third = mystring[2]  # this is an index of the 3rd element
print("letter indexed at 2 is", third)

eighth = mystring[7]  # this is an index of the 8th element
print("letter indexed at 7 is", eighth)
````

## `len()` function¶
- We have used `for/while` loops to count the elements in a sequence
- Python has a built-in `len` function that will give us the number of elements in a string.

   ```Python
      len(string)
   ```

````python
name = "Sir Galahad"
length = len( name )
print(f"{name} has {length} characters" )
````

````python
food = "spam"
length = len( food )
print(f"{food} has {length} characters" )
````

````python
print(f"The last element of {food} is {food[length]}" ) # this will give an error
````

````python
print(f"The last element of {food} is {food[length-1]}" )  # up to the length-1
````

<br>

> ##### We can also index from the end backwards (like the `range()` function)

````python
print( f"The last element of {food} is {food[length-1]}" )
````

## Looping through a string
> We have already looked at looping through a string with a `for` loop

<br>

#### A `for` loop using the `in` operator
* This will print each letter

````python
my_string = "eggs and spam"

for letter in my_string:
    print(letter)
````

<br>

#### We can make a `for` loop using the **index** instead
* When using the index value, we need to use `range` instead

````python
print("Using a for loop")

length = len( my_string )  # we can make length a variable

for n in range( length ):
    print( n, my_string[n] )
````

<br>

#### And a `while` loop using the **index** value

````python
print("Using a while loop")

length = len(my_string)  # we can make length a variable

i = 0

while i < length:
    print(i, my_string[i])
    
    i += 1
````

## String Slicing

<center><img src="https://i.postimg.cc/BZMCNyzv/middle-of-pie.jpg" width=200></center>

Python can pull out sections of a string (**"substrings"**).  This is called **string slicing**
* > **NOTE:** This can be confusing and takes practice to really grasp

<br>

#### The syntax is:

```Python
    string_name[start:stop:step]
```

This slice **starts** at the first number and goes up to **stop-1**.  

<br>

> **Similar to `range()`** stop and step are optional.  <br><br>
> If omitted, it is assumed you want the rest of the string one character at a time.

<br>

#### To print from the <u>beginning</u> use a 0 or leave blank (with a colon)

````python
string = "Sir Bedevere"
print( string[0:5] )     # from beginning to char 4 inclusive
print( string[5:] )      # another method
````

<br>

#### To print to the <u>end</u> use the index, `len()` or a blank (with a colon)

````python
string = "Sir Boors"
print( string[5:9] )            # only letters 4-8
print( string[5:] )  # letters 4 until the end
print( string[4:] )             # also 4 until the end
````

<br>

#### Print only <u>part</u> of the string

````python
string = "Sir Lancelot"
print( string[0:3] )   # only letters 0-2
print( string[5:] )    # only letters 4 to the end
print( string[4:7] )   # only letters 4-6
````

<br>

#### Regularly skip characters (the step value)

````python
string = "King Arthur"
print( string[0::3] )   # from beginning to end, skip by 3s
print( string[1::2] )  # start at 1, go to the end, and skip by 2s
````

## IMPORTANT: Strings are **IMMUTABLE**
- This is a fancy way of saying "unchangeable"
- There are certain data types in Python that **cannot be changed**
- Strings are one of these

<br>

#### Trying to change an individual character in the string gives an error

````python
string = "Sir Robin"
string[1] = "Y"     # try to change the character at index 1
print( string )
````

<br>

````python
# Using a new variable
string = "Sir Robin"
newstring = "M" + string[2:]  # add M to the front of string
print( newstring )
````

## Methods (special String Functions)
Methods are a special type of function that are performed using <font color="green"><b>dot notation</b></font> on the string object variable
```Python
    string_name.method()
```

##### We have <u>already</u> used the following **string methods**

```Python
    string_name.lower()       # all lowercase
    string_name.upper()       # all uppercase
```

##### But there are many many many more:

```Python
    string_name.capitalize()  # capitalize first character
    string_name.title()       # capitalize first words
    string_name.swapcase()    # swap lower and uppercases
    string_name.strip()       # remove white space from both ends of string

    string_name.find(letter)  # find a letter
    string_name.replace(a, b) # find a string and replace it in the original string
    string_name.count()       # count how many strings found in the original string
    string_name.split()       # split on a delimeter - space is the default

    string_name.isalnum()     # is it alphanumeric?
    string_name.isalpha()     # is it a letter?
    string_name.isdigit()     # is a numerical digit?
```

<br>

#### Examples of string methods in use

> **Format methods**

````python
# string method examples
my_string = "monTy PytHon and the quEst foR the hOly graIL"

print("Formatting")
print("----------------------")
print("  Original   ", my_string)
print("  lower()    ", my_string.lower())             
print("  upper()    ", my_string.upper())             
print("  capitlize()", my_string.capitalize())        # capitalize the first letter
print("  title()    ", my_string.title())             # capitalize every word
print("  swapcase() ", my_string.swapcase())       
````

> **Data check methods**

````python
str1 = "13a"
str2 = "spam"
str3 = "1784"
str4 = "17.84"

print("Checking Data")
print("----------------------")
print(f"  {str1:5}  isalnum()?    {str1.isalnum()}")   # is alphanumeric
print(f"  {str1:5}  isalpha()?    {str1.isalpha()}")   # only contains letters
print(f"  {str2:5}  isalpha()?    {str2.isalpha()}")   # only contains letters
print(f"  {str2:5}  isdigit()?    {str2.isdigit()}")   # only contains digits
print(f"  {str3:5}  isdigit()?    {str3.isdigit()}")   # only contains digits
print(f"  {str4:5}  isdigit()?    {str4.isdigit()}")   # only contains digits
````

> **Find and replace methods**

````python
print("Find and manipulate")
print("----------------------")
print("  find('P')          ", my_string.find("P"))          # first index where found       
print("  replace('on','**') ", my_string.replace("on","**")) # create new string replacing value
print("  count('a')         ", my_string.count("a"))         # count occurrences
print("  split()            ", my_string.split())            # split on character (default space)
````

> **Whitespace removal method**

````python
my_string = "    monTy PytHon and the quEst foR the hOly graIL   "

print("Removing Whitespace")
print("----------------------")
print(f"The string is '{my_string}'")
print(f"It contains {len(my_string)} characters")

print()
new_my_string = my_string.strip()          # remove whitespace from both ends of the string
print(f"After strip() the string is now '{new_my_string}'")
print(f"It contains {len(new_my_string)} characters")
````
